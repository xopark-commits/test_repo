## 12. 이상치 처리

아래 조건을 모두 만족하는 데이터만 남기세요.

- `loc`
- `&` 는 AND 조건

### 빈칸 힌트

```python
df = df.loc[
    (df["주문수량"] >= __) &
    (df["주문수량"] <= __) &
    (df["주문금액"] >= __) &
    (df["주문금액"] <= ______)
```
