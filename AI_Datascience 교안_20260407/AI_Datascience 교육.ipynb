## 파이썬 기본 문법

## Python VS Java/C++ 개념비교

🟩 Java / C++ : 소스 코드 → 컴파일 → 바이너리 → 실행 → OS/VM

    타입검사 : 컴파일 타임 
    오류 발견시점 : 빌드 시 

🟩 파이썬 : 소스 코드 → 한줄 씩 해석하며 실행(Interpereter)

    타입검사 : 실행시 (Runtime)
    오류 발견시점 : 실행 중 

`파이썬은 프로그램이 아니라 실행되는 스크립트에 가깝다`

## Python에서 객체(Object)란?

> 값(value) + 타입(type) + 행동(method)을 하나로 묶은 실행 단위다

💡 파이썬은 모든것이 객체다.

```
객체(Object)
 ├─ 타입(type)
 ├─ 값(value)
 └─ 동작(method)
```

예)

```python
x = 10
```

```
객체
 ├─ type: int
 ├─ value: 10
 └─ methods: +, -, *, ...

```

👉 개체이다 보니 메모리가 커진다.

| 언어           | 정수 하나        |
| ------------ | ------------ |
| C++ int      | 4 bytes      |
| Java Integer | ~16 bytes    |
| Python int   | ~28 bytes 이상 |

## 20. dunder 메서드

> dunder = double underscore


예시)
```python
__init__
__str__
__add__
__len__
```

공식 명칭은 Special method (특수 메서드)이고 앞에 언더바 두개가 붙은 메서드를 던더 메서드라고 부른다.

dunder 메서드는 Python 문법(연산자, 함수, 키워드)과 객체를 연결해 주는 훅(hook)이다.

```python
a + b

# 객체간의 연산을 실제로는 아래와 같이 메서드 호출로 통일한다.
a.__add__(b)
```

---

<div style="font-size:1.7rem;">⭐ 가장 중요한 dunder 메서드들</div>

1.  (__name__) - 테스트 코드 분리관련

파이썬 파일은 실행될 수도 있고, import 될 수도 있다.

| 실행 방식            | `__name__`   |
| ---------------- | ------------ |
| `python test.py` | `"__main__"` |
| `import test`    | `"test"`     |

왜 구분이 필요한가?

🚨 문제상황

```python
# mymodule.py
print("hello")

def add(a, b):
    return a + b
```

```python
import mymodule
```

❌ import했는데 "hello" 출력됨
→ 라이브러리로 쓰기 불가능


```python
def add(a, b):
    return a + b

if __name__ == "__main__":
    print(add(2, 3))

```

✅ 직접 실행할 때만 실행, import 시에는 실행 안됨.

1. __init__ – 생성자

```python
class User:
    def __init__(self, name):
        self.name = name
```

객체 생성 직후 자동 호출, Java의 constructor와 동일한 역할

```python
u = User("Alice")
# 내부적으로
# User.__init__(u, "Alice")
```

2. __str__ / __repr__ – 문자열 표현

| 메서드        | 용도         |

### 모집단과 표본

- ✅ 모집단 (Population): 모집단은 `연구나 조사의 대상이 되는 전체 개체 또는 집단`을 나타냅니다. 모집단은 특정한 특성이나 속성을 가진 모든 개체의 집합이며, 보통 연구자가 관심을 가지고 있는 전체 그룹입니다. 모집단은 매우 크거나 무한한 경우가 많기 때문에 직접적인 조사나 측정이 어려울 수 있습니다. 예를 들어, '모든 대학생의 키'나 '모든 고객의 구매 패턴'과 같은 개념적인 범위를 말합니다.

- ✅ 모수 (Parameter): 모수는 `모집단의 특성을 나타내는 숫자로, 특정한 수치적 값`으로 표현됩니다. 모수는 모집단의 통계적 특성을 요약하거나 설명하는 데 사용됩니다. 예를 들어, 모든 대학생의 키의 평균이나 표준편차, 모든 고객의 평균 구매 금액 등이 모수에 해당합니다. 모

- ✅ 표본(Sample) : 모집단에서 일부만 뽑은 작은 집단

    - 모집단 전체를 조사하는 것은 시간/비용/노력이 많이 들기 때문에,
    - 일부만 대표로 뽑아서 조사하는 것
    - 표본으로 얻은 결과를 가지고 모집단을 추론합니다 (→ 추론통계)

</br>

- ✅ 통계량(statistic) : 해당 모집단에서 추출한 표본(sample)을 이용해 만든 것으로 표본들의 함수

                         Ex) 표본 평균, 표본 분산, 표본 표준편차


<img src="img/population.png" />

- ✅ 표본분포(Sample Distribution) : 통계량의 확률분포

모집단이 어떤 특정한 분포를 따르고 있을 때 n개의 표본을 추출했다면 그 표본들은 모집단의 확률분포에 기반하여 추출된 것이라고 할 수 있습니다. 이 관점으로 보면 표본들은 확률표본(random sample)이고 통계량은 그 표본들의 함수이므로 당연히 특정한 확률 구조를 갖는 확률변수로 볼 수 있습니다. 이 통계량의 확률 구조를 표본분포라고 합니다.

평균이 $\mu $고 분산이 $\sigma^2 $인 정규 모집단에서 n개의 표본을 뽑았다고 하면, 표본평균 $\overline{X} $ 는 정규확률샘플들의 선형 결합이고 이는 위에서 언급한 정규분포의 특성으로 인해 정규분포를 따를 것입니다. 구체적으로 보면 다음과 같은 정규분포를 따릅니다.

<span style="font-size:150%">$$ 모집단의\, 분포가\, N(\mu, \sigma^2) 라면,\, \overline{X} \sim N(\mu, \frac{\sigma^2}{n}) $$</span>

### 표본추출 (Data Sampling)
출처 : https://blog.naver.com/deselorigin/220916376529

표본추출방법에는 크게 확률표본추출과 비확률표본추출의 두 가지 방법이 있고 확률표본추출법에는 단순임의추출법(simple random sampling), 층화임의추출법(stratified random sampling), 집락추출법(cluster sampling), 체계적 추출법(systematic sampling)의 네 가지 방법이 있습니다. 단순임의추출법은 가장 기본적인 방법으로 실제 조사에서 적용되는 경우는 드물다고 하겠습니다. 그 외 세 가지 방법들은 상호배타적으로 활용하는 것이 아니라 조합하여 다양한 표본추출방법의 적용이 가능합니다.

비확률표본추출에는 편의표본추출(convenience sampling), 판단표본추출(purposive sampling) 및 할당표본추출(quota sampling)이 있습니다. 조사설계의 실무에서 외견상 할당표본추출은 층화추출과 유사해 보입니다. 그러나, 실제 표본추출방법은 전혀 다르므로 두 가지는 준별해야 합니다.

#### 단순 임의 추출 (SRS: Simple Random Sampling)

<center><div style="background:white"><img src="img/random_sampling.png" /></div></center>

- **개념**: 모집단의 모든 개체가 **동일한 선택확률**을 갖도록 무작위로 `n`개를 뽑는 가장 기본적인 방법(복원/비복원 모두 가능).
- **절차**: 표본추출틀(frame)에 ID 부여 → 난수 생성(또는 `sample()`) → `n`개 선정.
- **추정량/불확실성**  
  - 모평균 추정치는 **표본평균** `ȳ` (불편).  
  - 표준오차(SE): `SE ≈ s * sqrt((1 − n/N)/n)` (유한모집단보정 포함).
- **장점**: 구현이 쉽고 가정이 최소. 통계적 성질(불편성·일치성)이 명확. 결과 해석이 직관적.
- **단점**: **완전한 프레임**이 필요하며, 표본이 지리적으로 흩어지면 조사비용↑.
- **잠재 오류**: 프레임 누락(coverage)·무응답(nonresponse) 발생 시 가중치/보정 필요.
- **데이터사이언스 관점**: 학습데이터에 **동일 가중치** 부여가 자연스러움. 신뢰구간·부트스트랩 보고가 용이.
- **예시**:
> 고객 10,000명 중 **무작위 400명**을 뽑아 NPS 조사. `ȳ=42`, `s=18`이면 `SE≈18*sqrt((1−0.04)/400)≈0.88`, 95% CI ≈ `42±1.96*0.88` → **[40.3, 43.7]**.
- **실무 팁**: 스트리밍 로그처럼 시간상관이 있으면 무작위 섞기 전 **블록화/셔플** 권장.
- **언제 쓰나**: 모집단 이질성이 크지 않고 프레임이 잘 준비되어 있을 때, 빠른 기준선이 필요할 때.

#### 층화 표본 추출 (Stratified Sampling)

<center><div style="background:white"><img src="img/stratified_sampling.png" /></div></center>

- **개념**: 모집단을 **이질적인 하위집단(층)**으로 나누고, 각 층 내는 **동질**하도록 구성한 뒤 층별로 SRS/체계적 추출을 수행.
- **목표**: 층간 차이를 통제해 **분산 감소**, **대표성 확보**, **소수층 보장**.
- **할당 방식**:  
  - **비례할당**: `n_h = n * (N_h/N)` (단순·직관).  
  - **네이만 최적할당**: `n_h ∝ N_h * S_h` (층 내 변동성이 큰 층에 더 배분 → 정밀도 극대화).
- **추정량/분산**: 전체 평균 `μ̂ = Σ_h (N_h/N) * ȳ_h`. 분산은 층별 분산의 가중합(유한모집단보정 적용).
- **장점**: 같은 표본크기에서 **정밀도↑**. 각 층에 대한 **부분 추정치**(예: 연령·지역)도 동시에 정확.
- **단점**: 정확한 **층 정보**가 사전 필요. 층 경계가 자주 바뀌면 운영비용↑.
- **데이터사이언스 관점**: 희귀 클래스(소수집단)를 **의도적으로 충분히 채집**해 모델이 소수 클래스를 학습하도록 설계. 추정 단계에서는 **층 가중치**로 전체 지표 보정.


- **예시**:  
>  통신사 가입자 100만 명을 **연령층(10대, 20대, …)**으로 나눔. 20·30대 변동성 `S_h`가 크면 네이만 할당으로 표본을 더 배분. 가입 만족도 평균 `μ̂`와 층별 평균/신뢰구간을 함께 보고.
>

- **실무 팁**: 분석 목적(추정 vs 분류)과 자원(예산·시간)에 맞춰 **층 정의/할당 규칙**을 문서화. 사후보정(post-stratification)로 가중치 재조정 가능.

#### 계통 추출 (Systematic Sampling)

<center><div style="background:white"><img src="img/systematic_sampling.png" /></div></center>

- **개념**: 간격 `k = ⌊N/n⌋`을 정하고 **1~k 중 임의 시작점 `r`**을 뽑은 뒤 `r, r+k, r+2k, …`로 표본화.
- **절차**: 정렬된 리스트에서 간격 추출. (리스트가 미리 무작위화되어 있으면 SRS와 근사 성질)
- **장점**: 구현이 매우 간단하고 **균등 분포**로 커버리지가 좋음. 현장조사 동선 효율↑.
- **단점**: 데이터 순서에 **주기성(periodicity)**이 있으면 **동조 편향** 발생(예: 매 60번째가 VIP인 정렬).
- **추정량/분산**: 표본평균을 모평균으로 사용(무작위 시작점 가정). 분산 추정은 순서상관에 민감하여 보수적으로 SRS 분산 사용하기도 함.
- **데이터사이언스 관점**: 로그/생산라인 등 **연속 흐름 데이터**에서 추출 속도가 빠르고 간단. 다만 **사전 랜덤화** 혹은 **주기성 점검** 필수.
- **예시**:
> 12,000건 결제 로그에서 `n=200`. `k=60`, 시작점 `r=17` → **17, 77, 137, …**번째 레코드 수집. 시간대 60분 주기 패턴이 있으면 먼저 셔플/블록화 후 시행.
- **실무 팁**: 대형 테이블은 `ROW_NUMBER() % k = r`로 구현해 인덱스 스캔 비용을 절감.
- **언제 쓰나**: 빠르고 균등한 커버가 필요하고 **주기성 위험이 낮거나 제거 가능**할 때.

#### 군집 추출 (Cluster Sampling)

<center><div style="background:white"><img src="img/cluster_sampling.png" /></div></center>

- **개념**: 모집단을 **자연적 묶음(군집: 학교·지점·가구 등)**으로 나누고, **일부 군집만 추출**하여 그 안의 모든(1단계) 또는 일부(2단계) 개체를 조사.
- **1단계 군집추출**: 군집을 무작위로 `m`개 뽑아 **군집 내 전수조사**. 간편하지만 군집 내 상관(ICC)↑이면 분산↑.
- **2단계(다단계)**: 군집을 뽑은 뒤 **군집 내에서 추가로 표본추출**(예: 반→학생 SRS). **비용 절감**과 **현장 동선 최적화**에 유리.
- **크기불균등 군집(PPS)**: 군집 크기가 제각각이면 **확률비례추출**로 군집을 뽑고, 최종 추정에서 **가중치**로 보정.
- **장점**: 프레임이 군집 단위로만 존재하거나 전국 단위처럼 **이동/조사 비용**이 큰 경우 매우 효율적.
- **단점**: **클러스터 내 유사성** 때문에 SRS 대비 **분산↑**(디자인 이펙트>1). 정확한 분산추정과 가중치 관리 필수.
- **추정량/가중치**: 1단계 전수는 **군집평균의 가중평균**(군집 크기 가중). 2단계는 각 단계 선택확률을 곱해 최종 **가중치 `w=1/π`**로 가중 추정.
- **데이터사이언스 관점**: 지역/지점 로그처럼 **물리적 접근 제약**이 있을 때 현실적. 모델 평가·지표 산출 시 **군집 가중/디자인 이펙트** 반영 권장.
- **예시**:
> 전국 편의점 12,000개 중 **무작위 300개 점포(군집)**를 뽑아 한 달간 모든 영수증을 수집(1단계). 또는 점포 300개를 뽑고 **일별 무작위 50건 영수증**만 수집(2단계). 최종 평균은 **군집·일 단위 가중치** 반영해 추정.
- **실무 팁**: 군집 내 상관을 완화하려면 **군집 수를 늘리고 군집당 표본을 줄이는** 설계가 유리. PPS·다단계 설계를 문서화해 재현성 확보.


### 복원 추출과 비복원 추출

#### 복원 추출 (Sampling With Replacement, WR)

 - 표본으로 뽑힌 개체를 **다시 모집단으로 되돌려** 다음 추출에 **다시 선택될 수 있게** 하는 방식.
  - 각 추출이 **독립**이며, 선택확률이 매번 동일(동일 분포)하다는 점이 특징.
  - 이론적으로 **무한모집단 가정**이나 대규모 모집단 근사에 적합.

#### 비복원 추출 (Sampling Without Replacement, WOR)

  - 한 번 뽑힌 개체는 **되돌리지 않음**. 동일 개체가 중복 선택되지 않는다.
  - 추출 간 **상관**이 생기며(선택 가능한 개체 수가 줄어듦) 모집단 크기를 명시적으로 반영.
  - **유한모집단**에서 더 현실적이며, 표본 간 중복을 피할 수 있어 **분산이 더 작아지는 경향**.

<center><div style="background:white"><img src="img/sampling_wr.png" /></div></center>

#### 통계적 성질과 차이

- **표본평균의 불편성**: 두 방식 모두 적절한 단순무작위 설계하에서 **표본평균 ȳ는 모평균 μ의 불편추정량**.
- **분산/표준오차(SE)**
  - 복원(W R): `Var(ȳ) = S² / n` → `SE = S / √n` (S: 모집단 표준편차 또는 표본표준편차로 근사)
  - 비복원(WOR): `Var(ȳ) = (S² / n) * (1 − n/N)`\
    → **유한모집단보정(FPC)** = `√(1 − n/N)`로 **SE가 더 작아짐**
  - `n/N`이 작으면(대규모 모집단에서 작은 표본) **두 방식의 차이는 미미**.
- **확률 모델**
  - 복원: 사건 수 카운트가 **이항분포**(Binomial)로 모델링되는 경우가 많음.
  - 비복원: 사건 수 카운트가 **초기하분포**(Hypergeometric)로 모델링됨.

### 확률

확률은 `사건이 발생할 가능성을 숫자로 나타내는 개념`

어떤 사건에 대해 어떤 결과가 나올지 확실하지 않는 상황일 때, 특정 결과가 나올 확률에 대해 얘기합니다.

각 확률을 토대로 사건에 대한 분석을 하는 학문을 통계학이라고 합니다.

<span style="font-size:200%">$$ 확률  = \frac{사건이 발생할 경우의 수}{전체 경우의 수} $$</span>
<span style="font-size:200%">$$ P(A)  = \frac{n(A)}{n(S)} $$</span>

<center>$ P(A) : 사건 A가 일어날 확률 $</center>
<center>$ n(A) : 사건 A가 일어날 수 있는 경우의 수 $</center>
<center>$ P(A) : 전체 경우의 수 $</center>

- ✅ 실험 (Experiment)

`실험은 어떤 과정이나 행위를 의미`합니다. 예를 들어, 동전 던지기, 주사위 던지기, 카드 뽑기 등이 실험의 예입니다. 
여기서 실험은 확률적 실험으로, 모든 가능한 결과(동전의 앞, 뒤면)를 알고 있어 반복 시행(100번)으로 규칙성(약 50번이 앞면이거나 뒷면)을 발견할 수 있는 실험을 의미합니다.

- ✅ 결과 (Outcome)

`실험을 수행한 후 얻을 수 있는 가능한 결과`들을 의미합니다.결과는 표본점(sample point) 또는 원소(element)라고도 불리며, 주사위를 던지는 실험에서 나올 수 있는 결과는 1, 2, 3, 4, 5, 6과 같은 것들입니다. 이 때 발생할 수 있는 모든 가능한 결과들의 집합 {1, 2, 3, 4, 5, 6}을 표본공간(sample space)라고 합니다. 이렇게 표본공간은 모든 가능한 결과들의 집합으로 이루어져 있으며, 결과들은 표본공간의 원소로 표현됩니다.


- ✅ 표본공간(sample space)

`실험에서 발생할 수 있는 모든 결과들의 집합을 표본 공간`이라고 합니다. 표본 공간은 대문자 S 나 Ω (오메가)로 표기됩니다. 예를 들어, 동전 던지기의 표본 공간은 {앞면(H), 뒷면(T)}로 표현됩니다.

- ✅ 사건(event)
  
`표본공간을 기반으로 우리가 관심을 가지는 사건(사상)`을 정의할 수 있습니다. 사건은 표본공간의 부분집합으로서, 특정 조건이나 기준에 따라 선택한 결과들의 집합입니다. 사건은 주로 알파벳 대문자로 표기됩니다. 예를 들어, A, B, C 등의 대문자로 사건을 나타낼 수 있습니다. 이 때, 각 사건은 표본 공간의 원소들을 포함하는 부분집합으로서 정의됩니다.

- ✅ 사건(event)
- 
경우의 수는 (말 그대로) `일어날 수 있는 가능한 "경우(상황)"의 총 개수`입니다.

예를 들어, 6면체 주사위를 던져 "홀수가 나오는 사건"을 생각해봅시다. 주사위의 모든 경우의 수는 6가지이며, 홀수인 경우는 1, 3, 5로 총 3가지입니다. 그리고 주사위를 던져서 발생할 수 있는 모든 가능한 결과들의 집합인 표본공간 S의 원소수, 홀수가 나오는 사건을 표본공간의 부분집합인 사건 A로 정의하고 수학적으로 n(S)=|S|=6, n(A)=|A|=3으로 표현합니다.

#### 확률 연산

- ✅ 곱의 법칙

: 상호 독립적인 사건(한 사건의 발생이 다른 사건의 발생에 영향을 주지 않는 경우를 말합니다. 예를 들어, 동전을 두 번 던진다고 가정해 봅시다. 첫 번째 던지기에서 앞면이 나올 확률과 두 번째 던지기에서 앞면이 나올 확률은 서로 독립적입니다. 첫 번째 던지기의 결과가 두 번째 던지기에 영향을 주지 않기 때문입니다.)의 경우, 각 사건이 발생할 수 있는 경우의 수를 곱하여 전체 경우의 수를 계산합니다. 예를 들어, 한 자리 수로 이루어진 3자리 수를 만들 때, 각 자리마다 가능한 경우의 수는 10가지이므로, 총 경우의 수는 10 x 10 x 10 = 1,000입니다.

<span style="font-size:200%">$$ P(A \cap B)  = P(A) \, \times \, P(B)  $$</span>

- ✅ 합의 법칙

: 상호 배타적인 사건의 경우, 각 사건이 발생할 수 있는 경우의 수를 합하여 전체 경우의 수를 계산합니다. 예를 들어, 한 자리 수로 이루어진 2자리 수를 만들 때, 첫 번째 자리 수가 1 또는 2일 경우, 두 번째 자리 수는 각각 10가지의 경우가 있으므로, 총 경우의 수는 10 + 10 = 20입니다.

<span style="font-size:200%">$$ P(A \cup B)  = P(A) \, + \, P(B) - P(A \cap B) $$</span>

#### 조건부 확률 (Conditional Probability)

- ✅ 조건부 확률(Conditional Probability)

:어떤 사건이 이미 발생했을 때, 그 조건 하에서 다른 사건이 일어날 확률을 의미합니다. 조건부 확률은 주어진 조건에 따라 사건이 발생할 가능성을 계산할 때 사용되며, 이렇게 특정 조건이 표본공간이 되는 확률입니다.

<span style="font-size:200%">$$ P(B | A)  =\frac{P(A \cap B)}{P(A)}  $$</span>

<center> <img src="img/cond_proba_define.png" /> </center>

<span style="font-size:200%">$$ P(A \cap B) = P(A) \cdot P(B|A) = P(B) \cdot P(A|B) $$</span>

- ✅ 독립성 (Independence)

: 두 사건 A와 B가 서로 독립적이라는 것은 하나의 사건이 일어났을 때 다른 사건에 대한 확률이 변하지 않는다는 것을 의미합니다. 즉, 사건 B가 일어났다고 해도 사건A가 일어날 확률에 아무런 영향을 미치지 않는 다는 것을 의미합니다. 따라서 
$P(A|B) = P(A)$ 와 $P(B|A) = P(B)$ 
가 성립합니다. 이때, 조건부 확률의 정의를 이용하여 독립적인 사건의 교집합의 확률이 각각의 확률을 곱한 값과 같다는 것을 다음과 같이 보일 수 있습니다.

<span style="font-size:200%">$$ P(A \cap B)  = 0 $$</span>
<span style="font-size:200%">$$ P(A \cup B)  = P(A) \, + \, P(B)  $$</span>

#### 전확률의 법칙 (Law of Total Probability)


어떤 사건 𝐴 의 전체 확률을, 사건들을 이루는 상호 배타적 분할 $(B_1,B_2,..,B_n)$을 이용해 구하는 방법입니다. 즉, 우리가 관심 있는 사건 𝐴 가 여러 경로(또는 조건)로 일어날 수 있을 때, 그 모든 가능한 경로를 더해서 전체 확률을 계산하는 법칙

<span style="font-size:200%">$$ P(B)  = \sum_i P(B|A_i) \cdot P(A_i) =  \sum_i P(B\cap A_i) $$</span>

#### 베이즈 정리 (Bayes' Theorem)

베이즈 정리를 정확히 배우기 전에 먼저 분할(Partition) 이라는 개념을 공부해야 합니다.  

K 개의 집합  $B_1,⋯,B_K$ 가 어떤 집합  S 의 분할이 되려면 두 조건을 만족해야 합니다.

> - 첫번째로  $ B_1,⋯,B_K$는 각각 서로소여야 합니다. 집합이 서로소 라는 것은 무작위로 두개를 뽑았을 때 겹치는 부분이 없어야 한다는 뜻입니다.
>
> - 두번째로  K 개의  $B_1,⋯,B_K$를 합집합 하였을때 그 집합은 정확히  S가 되어야 합니다.
  이 두 조건을 만족해야  B1,⋯,BK  를  S 의 분할이라고 할 수 있습니다.
>



<img src="img/bayes_define.png" />

자 이제  K 개의 집합  B1,⋯,BK 가 어떤 사건  S 의 분할이라고 합시다. 그러면 모든  S 의 부분집합  A 에 대해서 다음과 같은 식이 성립하게 됩니다.

<span style="font-size:200%">$$ A  = (A \cap B_1) \, \cup \, ... \, \cup \, ( A \cap B_K)   $$</span>

괄호안의 사건들이 각각 서로소이기 때문에 사건  A 가 일어날 확률은 아래의 식과 같이 계산 할 수 있습니다.
 

<span style="font-size:200%">$$ P(A)  = P(A \cap B_1) \, \cup \, ... \, \cup \, P( A \cap B_K)   $$</span>


<span style="font-size:200%">$$ P(B_j|A) = \frac{P(A|B_j)P(B_j)}{P(A)} = \frac{P(A|B_j)P(B_j)}{P(A|B_1)P(B_1)+...+P(A|B_K)P(B_K)}  $$</span>

### 기대값과 분산

---
#### 기대값(Expectation)

`기댓값은 확률 변수가 확률분포에 따라 갖게되는 값에 대한 평균적인 예상값`으로 확률분포의 중심(이 때 중심은 무게중심으로 중간값 median이 아닙니다.)을 나타냅니다. 

주어진 확률 변수의 가능한 모든 값에 대해 가중 평균(Weighted average)을 구하는 것으로 생각할 수 있습니다.

- 이산 확률 변수의 기댓값:

<span style="font-size:200%">$$ \mu = E(X) = \sum_x xP(X=x) $$</span>

또는 대신에 확률질량함수 $f(x)$ 로 표현합니다.

<span style="font-size:200%">$$ E(X) = \sum_x xf(x) $$</span>

- 연속 확률 변수의 기댓값:

<span style="font-size:200%">$$ \mu = E(X) = \int_{-\infty}^{\infty} xf(x)dx$$</span>

여기서 $ f(x)$는 X의 확률밀도함수이고 확률질량함수와는 달리 점확률 $P(X=x)$ 형태로 쓸 수 없습니다. 

- ✅ 기대값 성질

<span style="font-size:200%">$$ E(aX \pm bY) = aE(X) \pm bE(Y) $$</span>
<span style="font-size:200%">$$ E(aX \pm c) = E(X) \pm c $$</span>
<span style="font-size:200%">$$ E(XY) = E(X)\cdot E(Y) \quad (X,Y 가 \, 서로\, 독립일 때)$$ </span>
  

---
#### 분산(Variance)

`분산은 데이터의 흩어진 정도를 나타내는 통계적 측도`로, 데이터 값과 평균 간의 편차(차이)를 제곱한 후, 이 값을 평균하여 구합니다. 

데이터가 아닌 확률변수의 분산은 확률분포에서 확률변수의 변동성으로, 확률변수와 기댓값의 차이를 제곱한 값의 기댓값입니다.

<span style="font-size:200%">$$ \sigma^2 = Var(X) = E((X-E(X)^2) = E(X^2) - (E(X))^2$$</span>

분산은 제곱을 사용하기 때문에 원래 데이터 값과 단위가 다른 값이 됩니다. 

따라서 `같은 단위로 데이터의 변동성을 표현하기 위해서는 분산의 제곱근인 표준편차(Standard Deviation)를 사용`합니다.

<span style="font-size:200%">$$ \sigma = \sqrt{Var(X)} $$</span>

- ✅ 분산 성질

<span style="font-size:200%">$$ Var(X) \ge 0 → 편차 제곱합의 평균으로 양수 $$ </span>
<span style="font-size:200%">$$ Var(c) = 0 → 상수의 편차는 항상 \quad 0$$</span>
<span style="font-size:200%">$$ Var(cX) = c^2Var(X) → 상수배는 제곱으로 꺼낼 수 있으므로! $$</span>
<span style="font-size:200%">$$ Var(X \pm Y) = Var(X) + Var(X) (독립적인 \, 확률 변수의 합/차의 분산) $$</span>

### 확률 질량 함수

- ✅ 확률질량함수(probability mass function, PMF)
  
    확률 질량 함수는 이산 확률 변수에서 특정 값에 대한 확률을 나타내는 함수이다.

    즉, 이산 확률 변수의 확률구조를 나타낸다.


🟦 이산확률변수(discrete random variable) :

확률변수가 가질 수 있는 값들이 가산(countable) 또는 셀 수 있는 경우
예) 불량품의 개수, 사고 건수,...

🟦 표현방법

    - 확률변수 : P(X=1) (Probability의 P)
    
    - 확률질량함수 : f(1) ( function의 f)
$$ f(x) = P(X =x)$$

🟦 예제 : 동전 3개 던지기

    - 동전을 3개 던지는 확률실험을 할 때, 표본공간은 총 8개 였다.
    
    - 확률변수 X(앞면의 수) 의 값은 0, 1, 2, 3 으로 나타난다.

각 확률변수의 값에 대한 확률은 아래와 같이 나타난다.

$$  P(X =0) =  P({TTT}) = \frac{1}{8}$$
$$  P(X =1) =  P({HTT,THT,TTH}) = \frac{3}{8}$$
$$  P(X =2) =  P({HHT,HTH,THH}) = \frac{3}{8}$$
$$  P(X =3) =  P({HHH}) = \frac{1}{8}$$

확률질량함수로 표현하면,

$$  f(0) =  \frac{1}{8},f(1) =  \frac{3}{8},f(2) =  \frac{3}{8},f(3) = \frac{1}{8}$$

### 확률 밀도 함수

- ✅ 확률밀도함수(probability density function)

확률 밀도 함수는 연속 확률 변수의 분포를 나타내는 함수로 이산확률변수의 확률질량함수와 대응된다.

    - 가질 수 있는 값이 셀 수 없을 정도로 많은 경우
    - 예) 신장, 체중,..

🟦 히스토그램( Histogram )

    - 히스토그램은 수치자료 특히 연속자료의 분포형태를 표시
    - 계급의 상대도수를 사각형의 면적으로 표시 ⇒ 전체 면적 = 1

### 확률분포

`확률 분포는 확률 변수가 가질 수 있는 모든 값과 그 값들이 나타날 확률을 나타내는 함수`입니다. 

확률 분포는 확률 변수의 동작 방식을 묘사하며, 이를 통해 확률 변수의 특성을 이해하고 모집단의 특성을 추론할 수 있습니다.

주요한 확률 분포에는 정규 분포, 이항 분포, 포아송 분포 등이 있습니다. 

수학적으로 확률분포는 몇 개의 모수로 구성된 함수로 확률변수의 값을 정의역으로 갖습니다.

따라서 확률은 사건의 가능성을 수량적으로 나타내는 개념이고, 확률 분포는 확률 변수의 가능한 값과 그 값들의 확률을 나타내는 함수로서 모집단을 추정하거나 분석하는 통계적인 의미가 있습니다.

이산형 확률 변수는 다음과 같은 확률을 갖고, 

$$ \frac{사건 \,\, A가 일어날\, 개수}{전체\, 가능한\, 개수} $$

#### 이산확률분포 - 베르누이 분포

베르누이 분포는 `단일한 베르누이 시행의 결과를 나타내는 이산 확률 분포`입니다. 베르누이 시행은 매우 단순한 개념으로 `두 가지 가능한 결과 중 하나가 나오는 실험`을 말합니다. 실험 결과는 "성공"과 "실패"로 구분할 수 있으며(배반사건), 각각의 결과가 발생할 확률은 고정되어 있다고 가정합니다.

예를 들어, 동전 던지기는 베르누이 시행의 한 예입니다. 동전을 던졌을 때 앞면이 나올 경우를 "성공", 뒷면이 나올 경우를 "실패"로 정의

베르누이 분포의 확률질량함수는 다음과 같이 정의

<span style="font-size:200%">$$ f(x)=P(X=x)=Bern(p)=\begin{cases}p, & if\,\, x = 1\\1-p, & if\,\, x =0\end{cases} $$ </span>

여기서, p는 성공 확률이고 (1-p)는 실패 확률

예를 들어, 동전을 던져 앞면이 나올 확률을 p라고 할 때, 앞면이 나오면 X=1로 표시되고, 뒷면이 나오면 X=0으로 표시됩니다. 이때 베르누이 분포를 사용하여 앞면이 나올 확률을 계산하고 해당 확률을 기반으로 다양한 추론과 결정을 할 수 있습니다.

🟦 주요 특성 

기대값은 베르누이 분포에서는 성공1 과 실패0 두 가지 경우의 값이 있으므로 성공의 확률인 p와 실패의 확률인 1-p로 가중평균을 계산합니다.

<span style="font-size:200%">$$ E(X) = 1\cdot p + 0\cdot (1-p) = p $$ </span>

분산은 제곱의 평균 $1^2 \cdot p + 0^2 \cdot (1-p) $ 빼기 평균의 제곱 $p^2$ 으로 표현되며, p가 0.5일 때 최대값을 가지게 됩니다. p가 0 또는 1에 가까워질수록 분산은 작아지며, 성공 확률과 실패 확률이 균형을 이루는 경우에는 분산이 최대가 됩니다.

<span style="font-size:200%">$$ Var(X) = p- p^2 = p(1-p) $$ </span>

#### 이산확률분포 - 이항 분포

`이항 분포는 베르누이 시행을 독립적으로 반복하여 결과를 관찰하는 경우에 사용되는 확률 분포`입니다. 이항 분포는 이진(binary) 결과를 가지는 시행에서 성공 횟수 의 확률분포를 나타내는데 사용됩니다. 베르누이 분포에서 확률변수 X 가 성공 1, 실패 0 두 개의 값만을 갖는다면 이항 분포에서 확률변수 X는 성공의 횟수로 양의 정수를 갖습니다.

이항 분포의 확률 질량 함수는 다음과 같이 표현됩니다.

<span style="font-size:200%">$$ f(x) = P(X=x) = B(n,p) = \left(\begin{array}{c}n\\ x\end{array}\right) p^x (1-p)^{n-x}$$ </span>
<span style="font-size:200%">$$  = \frac{n!}{x!(n-x)!} \cdot p^x (1-p)^{n-x}$$ </span>

- $ P(X=x)는 x번 성공할 확률 $
- $ \left(\begin{array}{c}n\\ x\end{array}\right) $ 는 이항 계수(binomial coefficent)로, n개의 시행 중에서 x번의 성공이 발생할 수 있는 조합의 수
- $p,\,\, (1-p),\,\, n $는 각 성공확률, 실패확률, 시행 총 횟수 

</br>

🟦 주요 특성 

<span style="font-size:200%">$$ 기대값 : E(X) = np $$ </span>
<span style="font-size:200%">$$ 분산 : Var(X) = np(1-p) $$ </span>

#### 이산확률분포  - 포아송분포

`포아송 분포(Poisson distribution)는 단위 시간 또는 단위 공간에서 발생하는 사건의 수를 모델링하는 이산 확률 분포`입니다. 주로 사건이 독립적으로 발생하며, 발생률이 일정한 경우에 사용됩니다. 예를 들어, 단위 시간 동안 도착하는 이메일의 수, 단위 공간에서 발생하는 사고의 수, 특정 기간 동안 인터넷에서의 검색어 등장 횟수 등을 모델링할 때 포아송 분포를 사용할 수 있습니다.

포아송 분포는 이항 분포에서 시행 횟수 n이 충분히 크고 발생 확률p가 매우 작은 경우인 포아송 과정일 때 근사적으로 사용될 수 있습니다. 이 경우 이항 분포가 포아송 분포로 수렴하게 됩니다.`(n > 100, p < 0.01 이라면` $ \lambda = np $로 `근사화`시켜 풀수 있습니다. )

즉, 단위 시간이 어떤 시간을 매우 짧은 시간으로 나눈 구간이고 각 구간에서 똑같은 발생확률로 일어나는 독립적인 사건의 발생횟수는 포아송분포를 따릅니다.

$n \rightarrow \infty, \quad p \rightarrow 0, \quad np = \lambda $일 때, $p= \frac{\lambda}{n} $로 바꿔 이항분포질량함수에 $ n \rightarrow \infty$ 하면 포아송분포의 질량 함수가 됩니다.

<span style="font-size:200%">$$ \lim_{n \rightarrow \infty}\left(\begin{array}{c}n\\ x\end{array}\right) p^x (1-p)^{n-x} = \frac{e^{-\lambda}\lambda^x}{x!}$$ </span>

포아송 분포의 확률질량함수는 다음과 같이 표현됩니다.

<span style="font-size:200%">$$ f(x) = P(X=x) = poi(\lambda) = \frac{e^{-\lambda}\lambda^x}{x!}, \quad x = 0,1,2,..$$ </span>

🟦 포아송 분포 예

- 일정 주어진 시간 동안에 도착한 고객의 수
- 1킬로미터 도로에 있는 흠집의 수
- 일정 주어진 생산시간 동안 발생하는 불량 수
- 하룻동안 발생하는 출생자 수
- 어떤 시간 동안 톨게이트를 통과하는 차량의 수
- 어떤 페이지 하나를 완성하는데 발생하는 오타의 발생률
  
🟦 주요 특성 

<span style="font-size:200%">$$ 기대값 : E(X) = \lambda $$ </span>
<span style="font-size:200%">$$ 분산 : Var(X) = \lambda $$ </span>

#### 연속확률분포 - 균등분포(Uniform distribution)

`균등분포(Uniform distribution, 혹은 균일분포)는 주어진 구간에서 모든 값들이 동일한 확률로 나타날 때 사용`됩니다. 

특히 균등분포는 베이지안 통계에서 사전분포에 대한 정보가 전혀 없을 때, 무정보 사전분포로 사용됩니다.

<span style="font-size:200%">$$ f(x)=U(a,b) =\begin{cases}\frac{1}{b-a}, & if\,\, a \le x \le b\\0, &otherwise \end{cases} $$ </span>

여기서 a는 구간의 시작점이고 b는 구간의 끝점을 나타냅니다. 함수 f(x)는 구간 [a, b] 사이의 값들에 대해 동일한 확률을 할당합니다. 그 외 구간은 확률이 0입니다.

🟦 예시 : 0~10 사이의 무작위 시간

    - 어떤 사람이 0분에서 10분 사이에 걸려 출근한다고 가정.
    - “정확히” 3.742분에 도착할 확률은 0이지만,
    - 3~4분 구간에 들어올 확률은 균일하게 분포.


🟦 주요 특성 

<span style="font-size:200%">$$ 기대값 : E(X) = \frac{b+a}{2} $$ </span>
<span style="font-size:200%">$$ 분산 : Var(X) = \frac{(b-a)^2}{12} $$ </span>

#### 연속확률분포 - 정규분포(Normal distribution)

정규분포는 연속확률분포로서, `실수 전체의 범위에서 정의`됩니다. `두 개의 모수, 평균과 분산을 가지고 있으며 종모양의 대칭적인 형태`를 갖추고 있습니다. (이 모양은 대칭 형태와 확률의 총합 1을 유지하면서 분산에 따라 높이와 폭이 반비례하며 변합니다. 분산이 1인 표준정규분포의 경우 높이가 $ \frac{1}{\sqrt{2\pi}} \approx = 0.3989 $로 약 0.4이고 종의 밑부분의 폭이 4 정도에서 95%의 종모양이 완성되는 비율이기 때문에 오히려 납작한 스프 접시에 가깝다고 볼 수 있습니다. 물론 y축은 확률을 x축은 확률변수를 나타내기 때문에 서로 다른 단위입니다.)

정규분포의 PDF(확률밀도함수)는 다음과 같이 표현됩니다.

 <span style="font-size:200%"> $$ f(x) = N(\mu,\sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}}\cdot e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$ </span>

여기서, $x$는 확률변수, $\mu$는 평균, $\sigma^2$은 분산입니다.

🟦 주요 특성 

<span style="font-size:200%">$$ 기대값 : E(X) = \mu $$ </span>
<span style="font-size:200%">$$ 분산 : Var(X) = \sigma^2 $$ </span>

정규분포는 중심극한정리와 밀접한 관련이 있어서 많은 경우 실제 자료가 정규분포에 근사한다고 가정하고 분석합니다.

예를 들어 `이항분포의 경우 n이 충분히 크면 중심극한정리에 의해 평균 np, 분산np(1-p)를 갖는 정규분포를 따르게 됩니다.` 또한 정규분포는 자연 현상에서 많이 나타나는 형태로, 여러 요인의 합이나 평균으로 이루어진 데이터는 정규분포에 근사하게 됩니다. 이를 통해 모집단의 형태를 알지 못해도 모집단의 평균을 예측할 수 있으며, 그 예측값의 신뢰도를 수량화할 수 있습니다. 

<center> <img src="img/normal_graph.png" /> </center>

일반적으로 많이 쓰이는 확률값은 평규 좌우의 두, 세배 표준편차 구간의 확률입니다.

- $P(\mu-1\sigma \lt X \lt \mu+1\sigma) \approx 68  $ %
- $P(\mu-2\sigma \lt X \lt \mu+2\sigma) \approx 95  $ %
- $P(\mu-3\sigma \lt X \lt \mu+3\sigma) \approx 99.7  $ %

🟦 표준 정규 분포(Standard Normal Distribution)
 
 > 평균이 0이고 분산이 1인 정규 분포를 의미합니다.

X가 정규분포를 따르는 확률 변수일 때 $ X \sim N(\mu, \sigma^2)$ 표준 정규 분포로 변환된 Z는 다음과 같이 나타낼 수 있습니다

<span style="font-size:200%">$$ Z = \frac{X-\mu}{\sigma} $$ </span>

여기서 Z는 표준 정규 분포를 따르는 확률 변수를 의미합니다. 이렇게 표준 정규 분포로 변환된 Z 값은 평균이 0이고 분산이 1인 표준 정규 분포를 따릅니다. 

<span style="font-size:200%">$$ Z \sim N(0,1)$$ </span>

- $P(-1 \lt Z \lt +1) \approx 68.2  $ %
- $P(-2 \lt Z \lt +2) \approx 95.4  $ %
- $P(-3 \lt Z \lt +3) \approx 99.7  $ %

`여기 95%, 99% 해당 범위 숫자 1.96, 2.57 중요하니까 꼭 외워야됩니다.🔥`

- $P(-1.64 \lt Z \lt +1.64) \approx 90  $ %
- $P(-1.96 \lt Z \lt +1.96) \approx 95  $ %
- $P(-2.57 \lt Z \lt +2.57) \approx 99  $ %

🟦 중심극한정리 (Central Limit Theorem, CLT)
 > 확률론과 통계학에서 중요한 개념 중 하나로, 여러 독립적인 확률 변수의 합이 정규분포에 근사적으로 따른다는 원리

중심극한정리의 핵심 아이디어는 다음과 같습니다:


- 독립적인 확률 변수의 합: 중심극한정리는 독립적인 확률 변수들의 합에 대해 적용됩니다. 이러한 확률 변수들은 동일한 분포를 따를 필요는 없습니다.
- 표본 크기의 충분성: 표본 크기가 충분히 클 때, 확률 변수들의 합은 정규분포에 근사적으로 따릅니다. 통상적으로 표본 크기가 30 이상일 때 중심극한정리가 잘 작동합니다.

#### 연속확률분포 - 지수분포

지수분포는 `첫사건이 발생하는 데 걸리는 시간분포`입니다. 첫사건이란 사건이 일어날 횟수가 포아송분포를 따른다는 가정아래, 이 사건이 처음 일어나는데 걸리는 시간 또는 임의의 사건이 일어나고 다음 사건이 일어 나기 까지의 시간으로 이 시간은 지수분포를 따릅니다.

다시말해, λ가 서로 독립인 사건이 발생하는 평균 속도라면 이 때 사건이 발생할 `횟수`는 `포아송분포`를, 사건이 발생할 때까지의 `시간`은 `지수분포`를 따릅니다. 따라서 지수분포와 감마분포는 모두 모수로 λ를 갖습니다.

지수분포의 확률밀도함수는 다음과 같이 정의됩니다

<span style="font-size:200%">$$ f(x) =\begin{cases}\lambda e^{-\lambda x}, & x \ge 0\\0, &x \lt 0 \end{cases} $$ 
</span>

여기서 e는 자연상수(2.71828...)를 의미합니다.

$\lambda $는 평균속도이므로 $\frac{1}{\lambda} $ 시간 간격으로 일어나는 사건이 x시간 안에 일어날 확률은 누적분포함수로 표현됩니다.(간단한 예로, t시간 간격으로 일어나는 사건이 t시간 안에 일어날 확률은 1-0.37=0.63입니다.)

<span style="font-size:200%">$$  P(X \le x) = F(x) =\begin{cases} 1-e^{-\lambda x}, & x \ge 0\\0, &x \lt 0 \end{cases} $$ 
</span>

만약 사건이 x시간 이후에 일어날 확률이라면 다음과 같이 표현됩니다.(간단한 예로, t시간 간격으로 일어나는 사건이 t시간 이후에 일어날 확률은 0.37입니다.)

<span style="font-size:200%">$$  P(X \ge x) = 1 - F(x) =\begin{cases} e^{-\lambda x}, & x \ge 0\\0, &x \lt 0 \end{cases} $$ 
</span>

🟦 주요 특성 

- 지수분포는 무기억성(memoryless 분포
- 이전 사건이 발생한 시간과 관계없이 다음 사건이 발생하는 시간 간격은 독립적으로 분포
<span style="font-size:200%">$$ 기대값 : E(X) = \frac{1}{\lambda} $$ </span>
<span style="font-size:200%">$$ 분산 : Var(X) = \frac{1}{\lambda^2} $$ </span>

#### 연속확률분포 - t 분포
출처 : https://wikidocs.net/34009

분포를 쉽게 이해하려면, 이 분포는 순전히 `평균 검정을 하기 위해 고안되었다는 점`을 알고있어야 합니다.

만약 모집단이 정규분포를 따른다고 하면, 표본평균은 $ N(\mu, \frac{\sigma^2}{n}$을 따른다는 것을 기억하실 겁니다. 그래서 과거에는 이를 이용하여 평균 검정을 해왔는데, 일반적으로 우리는 `모분산인` $ \sigma^2 $을 `알 길이 없습니다.` 만약 n이 매우 크다면 표본평균은 더욱 정확히 정규분포를 따를 것이고, 표본평균의 분산 역시 0으로 점차 수렴하게 될 것입니다. 이 경우, 사실상 $ \sigma^2 $의 영향이 미미하게 되어 무시할 수 있지만, 표본 수가 작을 때는 문제가 될 수 있습니다. 모분산 $ \sigma^2 $을 정확히 알 수 없을 뿐 아니라, 그 값에 따라 정규분포의 모양이 크게 좌지우지되어, 정규분포를 이용한 검정이 그 신뢰성을 잃게되기 때문입니다.

`t분포는 정규분포와 같이 중심을 기준으로 좌우 대칭이고 종모양의 형태를 갖고 중심은 0으로 고정`되어있습니다. 즉, 위에서` 다룬 표준정규분포와 중심이 같고 자유도(degree of freedom, df)에 따라 종의 형태가 조금씩 변화`합니다. df는 표본 수와 관련이 있는 개념으로, 표본이 많아지면 표준정규분포와 거의 동일한 형태를 보입니다.

<span style="font-size:200%">$$ Y \sim t(n)이면,$$ </span>
<span style="font-size:200%">$$ f(y) =  \frac{\Gamma(\frac{n+1}{2})}{\Gamma(\frac{n}{2})\cdot\sqrt{\pi n}}\cdot(\frac{n}{y^2+n})^{\frac{n+1}{2}}, \quad -\infty \lt y \lt \infty $$ </span>

🟦 주요 특성 

<span style="font-size:200%">$$ E[Y] = 0$$ </span>
<span style="font-size:200%">$$ \quad V[Y] = \frac{n}{n-2}$$ </span>

t분포의 또 다른 특징은 `표본 수가 적으면 적을수록 표준정규분포에 비해 양쪽 꼬리가 더 두껍다는 것`입니다. `이는 데이터들이 그만큼 중심에 덜 모여있다는 것`입니다.

정규분포와 t분포는 검정의 관점에서 바라볼 필요도 있습니다. `정규분포를 이용한 검정의 경우, 중심에서 비교적 조금만 벗어나도 다르다는 결과를 보이지만 t분포는 정규분포에 비해 상대적으로 더 벗어나도 같다는 결과를 줍니다.` 이는 표본 수가 적은데서 나오는 우연에 의한 극단적인 값에 대해서도 어느정도 유연한 검정 결과를 준다고 할 수 있습니다.

🟦 T 분포 정리

- 정규분포와 다르게 t분포는 모분산 $\sigma^2 $를 포함하고 있지 않다. 이는 표본 수가 적을 때 신뢰성을 더해준다.
- 표본 수가 적을 때, t분포는 표준정규분포보다 양쪽 꼬리가 더 두텁다. 이는 보다 보수적인 검정을 하게 해준다.
- 표본수가 많아지면 t분포와 표준정규분포에 근사한다.

#### 연속확률분포 - 카이제곱분포

`통계에서는 변동(분산)은 비교, 관계 등 모든 분석에서 사용되는 개념`입니다. 변동은 단순히 해당 변수의 퍼짐 정도를 나타내는 것뿐 아니라 비교의 신뢰성을 표현하기도 합니다. 예를 들어, 두 집단의 평균이 큰 차이가 없어도 그 평균의 변동이 크다면 신뢰하기 힘듭니다. 또한 변동은 볼륨을 뜻하기도 합니다. A라는 변동과 B라는 변동이 있을 때, 비교를 통하여 더 큰 볼륨(파워, 확장성)을 가진 변동을 알 수 있습니다. 우리는 이러한 `변동을 측정하는 도구로 제곱합 구조` ( $\sum a^2 $) `사용`하게 됩니다. 

그 이유는 변동을 측정하는 방법에 있습니다. 

변동은 어떤 값을 기준으로 각 관찰값들이 그 기준값과 얼만큼 떨어져 있는지 평가하는 방식으로 측정하게 되는데, 대부분의 경우 평균을 사용하게 됩니다. 그렇지만 `단순히 평균과 관찰값들의 거리를 이용하면 그 합이 0이 되어 의미가 없어지기 때문`이죠. `이 구조적인 문제를 해결하고자 기준값과의 거리의 제곱을 이용`하게 됩니다. 그렇기에 변동은 오로지 양적인 값으로만 표현됩니다.

또한 두 집단의 변동을 비교하는 방법으로 두 변동의 차이를 사용하면 곤란합니다. `비교하려는 집단의 단위가 다를 수 있을 뿐더러 각 집단의 수를 반영해주어야 하기 때문`입니다. 예컨대, 한 집단은 cm 단위로 조사되었고 다른 집단은 m단위로 표현되어 있는 길이를 생각해 봅시다. 같은 수준의 변동을 가졌다고 해도 막상 제곱합을 구해보면 cm로 조사된 집단의 더 크게 나올 것입니다. 단위가 다르기 때문이죠. 또한 10개 데이터에서와 100개 데이터에서의 분산을 똑같이 보면 곤란합니다. 이는 '제곱의 합' 꼴로 표현되므로 자료 수가 많으면 많을수록 더욱 커질 수 밖에 없는 구조이기 때문이죠.

이러한 특성들 때문에, 변동을 이용하여 검정, 추정을 하고 싶으면 위와 같은 상황들을 전부 고려한 확률구조가 필요합니다. `변동의 '단위' 와 '자료 수' 를 반영할 수 있도록 고안된 분포`가 바로 $\chi^2 $분포이고  들의 비(ratio)가 따르는 분포가 $F$ 분포입니다. 
 $\chi^2 $ 들의 비는 두 변동을 비교하기 위해 사용되는 것으로, 이 역시 자료 수와 관련된 자유도가 반영됩니다.

$\chi^2 $ 변수를 만드는 아이디어는 정규분포를 따르는 변수에서 시작합니다. 우선 단위를 일반화하기 위해 표준화를 한 후 제곱을 하면 자유도 1인 $\chi^2 $ 변수가 됩니다. 또한 $\chi^2 $ 변수는 가법성이라고 불리는 성질이 있어, 독립적인 $\chi^2 $ 변수끼리 더해도 
$\chi^2 $ 분포를 따르게 됩니다.

<span style="font-size:200%">$$ Z \sim N(0,1) \quad \Rightarrow Z^2 \sim \chi^2(df =1)\ $$ </span>
<span style="font-size:200%">$$ \Rightarrow \sum_{i=1}^{n} Z_{i}^2 \sim \chi^2(df=n) \quad (Z_i 끼리는 \, 서로 \, 독립) $$ </span>

카이제곱분포의 PDF는 다음과 같이 정의됩니다.

<span style="font-size:200%">$$ f(x) = \chi^2(\nu) = \frac{x^{\frac{\nu}{2} -1}e^{-\frac{x}{2}} }{2^{\frac{\nu}{2}}\Gamma(\frac{\nu}{2})} $$ </span>

여기서, x는 확률변수의 값이며, $\nu $는 자유도(df:degree of freedom)를 나타냅니다


🟦 주요 특성 

<span style="font-size:200%">$$ E[X] = \nu $$ </span>
<span style="font-size:200%">$$  Var[X] = 2\nu$$ </span>

- 카이제곱분포는 `항상 양수`이며, `오른쪽으로 꼬리가 긴 형태`를 가집니다.
- 자유도에 따라 분포의 형태가 변화합니다. `자유도는 표본의 개수`에 의해 결정됩니다.
- 자유도가 작을수록 분포의 분산이 커지며, 자유도가 증가할수록 분산이 작아집니다.
- 카이제곱분포는 `주로 분산 추정, 적합도 검정(Goodness-of-Fit Test), 교차 분석(Cross Tabulation Analysis) 등의 통계적 분석`에서 사용됩니다.

#### 연속확률분포 - F 분포
출처 : https://blog.naver.com/psychometrics_/221518674493

F분포는 독립적인 두 카이 제곱 분포에 관한 비로써 정의하고 `2개 이상의 집단의 평균차이를 검정하는 경우나 두 분산의 차이를 검정할 때 사용하는 분포`이다.

$ U_1 \sim \chi_{\nu_1^2}$, $ U_2 \sim \chi_{\nu_2^2}$ 이고  $U_1 $, $U_2$ 가 서로 독립 일때 , F분포는 다음과 같이 정의됩니다.

<span style="font-size:200%">$$ X = \frac{U_1 / \nu_1}{U_2 / \nu_2} \sim F_{\nu_1,\nu_2}$$ </span>

$\nu_1 $은 $U_1 $(분자)의 자유도이고, $\nu_2 $은 $U_2 $(분자)의 자유도이다.

2개의 집단 간 평균차이는 쉽게 검정할 수 있지만, 만약 집단이 3개 이상으로 늘어난다면 여러 번의 t검정이 필요하다.

이렇게 되면 유의수준이 우리가 설정한 것보다 커지는 문제가 발생한다.

다시 말해, 세 집단의 평균차이를 검정한다면 영가설을 채택할 수 있는 확률이 아래와 같이 바뀐다.

한 번의 검정은 95%지만, 세 번의 검정을 실시한다면 85.7%가 된다.

$\left(1-0.05\right)\times \left(1-0.05\right)\times \left(1-0.05\right)\ \risingdotseq \ 0.857$

🟦 주요 특성 

<span style="font-size:200%">$$ E[X] = \frac{nu_2}{\nu_2 -2}(\nu_2 \gt 2)$$ </span>
<span style="font-size:200%">$$  Var[X] = \frac{2 \nu_{2}^2(\nu_1 + \nu_2 -2)}{\nu_1(\nu_2 -2)^2(\nu_2 -4)}(\nu_2 \gt 4)$$ </span>
<span style="font-size:200%">$$  f_{1-\alpha}(\nu_1, \nu_2) = \frac{1}{f_{\alpha}(\nu_2,\nu_1)}$$ </span>

### 데이터 스케일링 (Data Scaling)
출처 : https://na0-0.tistory.com/137


`데이터의 값의 범위를 조정하는 것, 데이터 수치를 일정한 수준으로 변환하는 것`

머신러닝을 위한 데이터셋을 정제할 때, 특성별로 데이터의 스케일이 다르다면 어떤 일이 벌어질까요?

예를 들어,

- X1은 0 부터 1 사이의 값을 갖고
- X2 는 1000000 부터 1000000000000 사이의 값을 갖고
- y 는 1000000 부터 100000000 사이의 값을 갖는다고 가정한다면
- X1 특성은 y 를 예측하는데 큰 영향을 주지 않는 것으로 생각할 수 있습니다.
  
때문에 특성별로 데이터의 스케일이 다르다면, 머신러닝이 잘 동작하지 않을 수 있습니다.

따라서, 저희는 데이터 스케일링 작업을 통해, 모든 특성의 범위(또는 분포)를 같게 만들어줘야합니다.

</br>

🟦 주로 사용되는 스케일링 개념

- Standardization (표준화)
> 특성들의 평균을 0, 분산을 1 로 스케일링하는 것입니다.
즉, 특성들을 정규분포로 만드는 것입니다.
- Normalization (정규화)
> 특성들을 특정 범위(주로 [0,1]) 로 스케일링 하는 것입니다.
가작 작은 값은 0, 가장 큰 값은 1 로 변환되므로, 모든 특성들은 [0, 1] 범위를 갖게됩니다.

 <center><div style="background:white" > <img src="img/scale_graph.png" /> </div></center>

</br>

🟦 스케일링이 필요한 경우

1. `주성분 분석(PCA)`

: 높은 분산/ 넓은 범위를 가지는 변수는 낮은 분산을 가지는 변수보다 주성분에서 큰 회귀계수를 가지게 된다. 즉, 중요하지 않은 변수도 범위가 넓으면 주성분에서 중요한 변수로 간주

2. `군집(clustering)`

: 거리-기반 알고리즘으로 거리 측도(ex. 유클리디안, 맨하튼 ... )를 이용하여 관측값 사이에서 유사성을 찾아 군집을 형성한다. 따라서, 넓은 범위를 가지는 변수는 군집에서 더 큰 영향력을 가진다.

 

3. `k-최근접 이웃(knn)`

: k-최근접 이웃은 거리-기반 알고리즘으로 새로운 관측값의 주변에 있는 k개의 이웃(데이터)를 이용하여 유사성 측도에 기반해 관측값을 분류한다. 따라서, 유사성 측도에 모든 변수가 동일하게 기여할 수 있도록 스케일링을 수행해야한다.


4. `서포트벡터머신(SVM)`

<center><div style="background:white" > <img src="img/scale_svm.png" /> </div></center>
</br>

: 서포트벡터 머신은 서포트벡터와 분류기(hyperplane) 사이 거리인 마진(margin)을 최대로 만들어주는 분류기를 찾는 알고리즘이다. 따라서, 특성 값이 크면 거리를 계산할 때 다른 특성보다 우선시 되어 영향력을 많이 미치게 된다.

 

5. `회귀모델에서 변수 중요도를 측정하기 전`

: 예측을 위한 회귀 모델링이 목적이라면 표준화를 수행할 필요가 없지만, 변수의 중요도나 다른 회귀계수들과 비교를 하는 것이 목적이라면 표준화가 수행된 독립변수를 이용하여 회귀 모형을 피팅하여 표준화된 계수의 절대값을 비교하여 회귀 분석에서의 변수 중요도를 측정할 수 있다.

 

6. `LASSO/RIDGE 회귀 전`

: 각 변수에 관련된 계수의 크기에 패널티를 부여하며, 변수의 규모는 계수에 적용되는 패널티 정도에 영향을 미친다. 분산이 큰 변수의 계수는 작아 불이익을 덜 받는다. 따라서 구 회귀를 모두 맞추려면 표준화가 필요하다.

🟦 스케일링이 필요하지 않은 경우

로지스틱 회귀, 트리 기반 모델(ex. 의사결정나무, 랜덤 포레스트 ...), 그래디언트 부스팅은 변수의 크기에 민감하지 않으므로 표준화를 수행해줄 필요가 없다.

#### StandardScaler

StandardScaler는 모든 피처들을 평균이 0, 분산이 1인 정규분포를 갖도록 만들어줍니다.

즉, 표준화해주는 방법이죠.

이 방법의 경우 `데이터 내에 이상치가 있다면 데이터의 평균과 분산에 크게 영향을 주기 때문에` 스케일링 방법으로 적절하지 않습니다.

<span style="font-size:200%"> $$ Z = \frac{X_i - mean(X)}{std(X)}
$$</span>

- 기존 변수에 범위를 정규 분포로 변환
- (x - x의 평균값) / (x의 표준편차)
- 데이터의 최소, 최대 값을 모를 경우 사용

````python
from sklearn.preprocessing import StandardScaler

std = StandardScaler()
std.fit(X_train)
X_train_scaled = std.transform(X_train)
X_test_scaled = std.transform(X_test)
dtc.fit(X_train_scaled, y_train)
print('모델의 정확도 :', round(dtc.score(X_test_scaled, y_test), 4))
````

#### MinMaxScaler

MinMaxScaler는 `모든 피처들이 0과 1사이의 데이터값을 갖도록` 만들어줍니다.

즉, 피처별로 최솟값은 0이 되고, 최댓값은 1이 되는 것이죠.

데이터가 2차원인 겅우, 모든 데이터는 x, y 축의 0과 1 사이에 존재하게 됩니다.

이 방법 또한 이상치가 존재한다면, 이상치가 극값이 되어 데이터가 아주 좁은 범위에 분포하게 되기 때문에 스케일링 방법으로 적절하지 않습니다.

<span style="font-size:200%"> $$ Z = \frac{X_i - min(X)}{max(X) -  min(X)}
$$</span>

- 데이터를 0-1사이의 값으로 변환
- (x - x의 최소값) / (x의 최대값 - x의 최소값)
- 데이터의 최소, 최대 값을 알 경우 사용


---
#### RobustScaler

<span style="font-size:200%"> $$ Z = \frac{X_i - Q_2(X)}{Q_3(X) -  Q_4(X)}
$$</span>

- StandardScaler에 의한 표준화보다 동일한 값을 더 넓게 분포
- `이상치(outlier)를 포함하는 데이터를 표준화하는 경우(영향을 최소화)`


데이터의 중앙값(median) = 0, IQE(interquartile range) = 1이 되도록 스케일링

RobustScaler를 사용할 경우 StandardScaler에 비해 스케일링 결과가 더 넓은 범위로 분포

#### MaxAbsScaler

최대 절대값과 0이 각 1, 0이 되도록 하여 양수 데이터로만 구성되게 스케일링

 <span style="font-size:200%"> $$ X_scaled = \frac{x}{max(|x|)}
$$</span>

데이터가 -1과 1 사이에 위치하도록 스케일링( 절대값의 최소값 = 0, 절대값의 최대값 = 1 )

데이터의 값이 양수만 존재할 경우 MinMaxScaler와 유사하게 동작

#### 위치 통계량 (평균, 중앙값, 분위수)

✅ 평균 

- 산술 평균 : 데이터의 합계를 데이터의 총 개수로 나눈값

 <span style="font-size:200%"> $$ \overline{x} = \frac{x_1 + x_2 + \cdot \cdot \cdot + x_n}{n}$$</span>

- 가중 평균 :산술평균에서 각 원소의 가중치가 다름

<span style="font-size:200%"> $$ \overline{x} = \frac{w_1 x_1+ w_2 x_2+ \cdot \cdot \cdot + w_n x_n}{w_1 +w_2 + \cdot \cdot \cdot +w_n }$$</span>

- 기하 평균 : 곱의 평균이며 변화량(성장률)의 평균을 계산하는데 주로 사용

<span style="font-size:200%"> $$ \overline{x_g} = \sqrt[n]{x_1\cdot x_2\cdot  x_3\cdot \cdot \cdot x_n} $$</span>

- 조화 평균 : 역수의 산술평균의 역수

<span style="font-size:200%"> $$ \overline{x_n} = \frac{n}{\frac{1}{x_1}+\frac{1}{x_2}+\frac{1}{x_3}+ \cdot \cdot \cdot +\frac{1}{x_n}}  $$</span>

✅ 중앙값 : 주어진 숫자 중에 가장 가운데 위치한 값 , 매우 크거나 작은 `이상치에 영향을 덜 받음`

- 숫자개수가 홀수인 경우

  > [ example ] 2,4,5,6,8 → 중앙값 5 

- 숫자개수가 짝수인 경우

  > [ example ] 2,4,6,8 → 중앙에 가까운 두값 4, 6의 평균 5가 중앙값


✅ 분위수 : 데이터 표본을 4개의 동일한 부분으로 나눈 값

사분위 범위(IQR, Inter Quantile Range): 제 3사분위수(3Q, 75%)와 제 1사분위수(1Q,25%)의 범위

> IQR = 3Q-1Q

#### 변이 통계량 (분산,표준편차,변동계수)

✅ 분산 : 자료의 퍼짐 정도를 확인하는 통계량

<center><div style="background:white" > <img src="img/var_graph2.png" /> </div></center>

> 편차 : $ (x - \overline{x})$
> 
> 편차의 제곱 : $ (x - \overline{x})^2$
> 
> 편차의 제곱 : $ \sum{(x - \overline{x})^2}$

<span style="font-size:200%"> $$ var = \frac{\sum{(x - \overline{x})^2}}{n} $$</span>

<span style="font-size:200%"> $$ var = \frac{\sum{(x - \overline{x})^2}}{n} $$</span>

✅ 표준편차 : 분산의 제곱근, 원래 데이터와 같은 단위라 해석이 직관적

<span style="font-size:200%"> $$ std = \sqrt{\frac{\sum{(x - \overline{x})^2}}{n}} $$</span>

✅ 변동계수 : 표준편차의 평균에 대한 비율

> 무명수(단위가 없는 수)로 서로 다른 단위를 가지는 변량 비교시 활용 가능

<span style="font-size:200%"> $$ CV = \frac{\sigma}{\mu} \times 100 $$</span>

$ \mu$ : 모평균(모집단의 평균)

$ \sigma$ : 모표준편차(모집단의 표준편차)



#### 모양 통계량 (첨도, 왜도)

✅ 첨도(kurtosis) : 분포의 뾰족한 정도를 나타내는 척도로 정규분포 기준 0이며 정규분포보다 뾰족하면 양수

<center><div style="background:white" > <img src="img/kurt_graph.png" /> </div></center>

<span style="font-size:200%"> $$ 첨도 = \frac{1}{n-1}\sum_{i=1}^{n}(\frac{x_i - x}{s})^4 -3 $$</span>

</br>

✅ 왜도(skewness) : 분포의 치우침 정도를 나타내는 척도로 좌우 대칭인 경우 0이며 오른쪽 고리 분포면 양수

- 데이터 분포의 비대칭성을 나타냄

##### ✅ 산점도

✔ 개념 정의

- 두 변수 간의 **관계(상관성)**을 시각적으로 보여주는 가장 기본적인 이변량 차트.
- 각 데이터 포인트는 “X값, Y값” 한 쌍으로 이루어진 **점(point)**으로 표현됨.
- X축: 독립변수(원인·특징), Y축: 종속변수(결과·반응)로 설정하는 것이 일반적.
- 산점도는 데이터 분포의 **형태, 패턴, 군집, 이상치(outlier)**를 한눈에 파악하는 데 매우 효과적.


<center><img src="https://otexts.com/fppkr/fpp_files/figure-html/corr-1.png" /></center>

✔ 산점도가 보여주는 핵심 정보

상관관계 방향

- 양의 상관: 오른쪽 위로 증가하는 패턴

- 음의 상관: 오른쪽 아래로 감소하는 패턴

상관 없음: 점이 무작위로 흩어짐

상관성의 강도

- 직선에 가까울수록 강한 상관

- 넓게 퍼질수록 약한 상관

군집(cluster) 존재 여부

- 여러 그룹이 생긴다면 군집 구조를 가진 데이터

이상치(outlier)

- 주요 패턴에서 멀리 떨어진 점 → 데이터 오류 또는 극단적 사례일 가능성


✔ 산점도의 장점

- 직관적이고 해석이 쉬움

- 두 변수의 관계를 직접 시각적으로 보여줌

- 이상치 탐지에 강함

- 선형·비선형 관계 모두 파악 가능

- 데이터의 전체적인 구조(structure)를 한눈에 이해

✔ 산점도의 단점

- 데이터가 수천~수만 개 이상이면 점들이 겹쳐 보이는 과포화(overplotting) 발생

- 카테고리 변수(범주형 변수)의 표현에는 적합하지 않음

- 해석에 주관적 요소가 약간 포함됨 (경향이 눈으로 보이는 패턴에 의존)

✔ 실생활/실무 예시

- 키(X)와 몸무게(Y)의 관계

- 제품 가격(X)과 판매량(Y)

- 온도(X)와 전력 사용량(Y)

### Part 1. Matplotlib 기본 그래프

#### 1-1. 가장 기본적인 선 그래프

`plt.plot()`은 Matplotlib에서 선 그래프를 그리는 가장 기본적인 함수입니다. 

리스트나 배열 형태의 숫자 데이터를 입력받아, 각 점을 순서대로 선으로 연결합니다. 

만약 하나의 데이터 리스트만 입력하면, 이 값들은 y축의 값이 되며, x축의 값은 자동으로 0부터 시작하는 정수 인덱스가 됩니다. 

시계열 데이터나 순차적인 데이터의 변화 추세를 시각화하는 데 가장 기본이 되는 형태입니다.

````python
plt.plot([1, 4, 9, 16])
plt.show()
````

#### 1-2. x, y 값을 모두 입력하여 그리기

`plt.plot()` 함수에 두 개의 리스트를 입력하면, 첫 번째 리스트는 x축의 값으로, 두 번째 리스트는 y축의 값으로 매핑됩니다.

즉, (x[0], y[0]), (x[1], y[1]), ... 와 같이 좌표 쌍을 이루는 점들을 생성하고 이 점들을 선으로 연결합니다. 

두 변수 간의 관계를 표현하고자 할 때 이처럼 x, y 값을 명시적으로 지정해야 합니다. 

두 리스트의 길이는 반드시 동일해야 합니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.show()
````

#### 1-3. 여러 선 그래프 그리기

하나의 그래프 영역(Axes) 안에 여러 개의 선을 동시에 표시하여 데이터를 비교 분석할 수 있습니다.

`plt.show()`가 호출되기 전까지 `plt.plot()` 함수를 여러 번 사용하면, 모든 선이 같은 그래프 영역에 겹쳐서 그려집니다.

각 선이 어떤 데이터를 나타내는지 구분하기 위해, `plot()` 함수 내에 `label` 인자를 지정하고 마지막에 `plt.legend()` 함수를 호출하여 범례를 표시하는 것이 좋습니다.

````python
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16] # 첫 번째 데이터
y2 = [1, 2, 4, 8]  # 두 번째 데이터

plt.plot(x, y1, label='A 데이터')
plt.plot(x, y2, label='B 데이터')
plt.title('두 종류 데이터 표현')
plt.xlabel('X축')
plt.ylabel('Y축')
plt.legend()
plt.show()
````

### Part 2. 그래프 기본 요소 설정

#### 2-1. 제목(Title) 추가하기

그래프의 제목은 해당 시각화가 무엇을 나타내는지 알려주는 가장 중요한 요소입니다. 

`plt.title()` 함수를 사용하여 그래프 상단에 텍스트를 추가할 수 있습니다. 

제목을 통해 독자는 그래프의 주제를 즉시 파악할 수 있으므로, 간결하고 명확하게 작성하는 것이 중요합니다. 

제목이 없는 그래프는 그 의미를 해석하기 어렵기 때문에 항상 추가하는 습관을 들이는 것이 좋습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('간단한 선 그래프')
plt.show()
````

##### 제목 위치와 정렬 조정하기

`plt.title()` 함수의 `loc` 파라미터를 사용하면 제목의 위치를 정렬할 수 있습니다.

`loc`의 값으로 'left', 'center', 'right'를 설정할 수 있으며, 기본값은 'center'입니다. 

이를 통해 그래프의 전체적인 디자인 균형에 맞춰 제목을 배치할 수 있습니다. 

`pad` 파라미터는 그래프 영역과 제목 사이의 간격(padding)을 조절하는 데 사용됩니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('왼쪽 정렬 제목', loc='left', pad=10)
plt.show()
````

#### 2-2. 축 레이블(Label) 추가하기

각 축이 어떤 데이터를 나타내는지 설명하는 축 레이블은 그래프를 해석하는 데 필수적입니다.

`plt.xlabel()`과 `plt.ylabel()` 함수를 사용하여 각각 x축과 y축에 대한 설명을 텍스트로 추가할 수 있습니다.

데이터의 단위(예: '시간(초)', '매출(억원)')를 함께 표기해주면 정보를 더욱 명확하게 전달할 수 있습니다. 

축 레이블이 없으면 각 축의 의미를 알 수 없어 그래프를 이해하기 어렵습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('축 레이블 추가')
plt.xlabel('X축')
plt.ylabel('Y축')
plt.show()
````

#### 2-3. 범례(Legend) 추가하기

하나의 그래프에 여러 개의 데이터 계열(선, 막대 등)이 있을 때, 각 계열이 무엇을 의미하는지 설명해주는 역할을 범례가 담당합니다. 

범례를 표시하려면, 먼저 `plt.plot()` 함수 내에서 `label` 파라미터로 각 선의 이름을 지정해주어야 합니다.

그 후, `plt.legend()` 함수를 호출하면 그래프의 빈 공간에 자동으로 범례가 생성되어 각 선의 스타일과 지정된 이름이 표시됩니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y, label='데이터')
plt.title('범례 추가')
plt.xlabel('X축')
plt.ylabel('Y축')
plt.legend()
plt.show()
````

##### 범례 위치 조정하기

범례가 데이터 포인트를 가리는 등 기본 위치가 적절하지 않을 때가 있습니다.

`plt.legend()` 함수의 `loc` 파라미터를 사용하면 범례의 위치를 수동으로 지정할 수 있습니다. 

`loc`에 올 수 있는 값은 'best'(기본값, 자동으로 최적 위치 선정), 'upper left', 'upper right', 'lower left', 'lower right', 'center' 등 다양합니다. 

이를 통해 그래프의 가독성을 높일 수 있습니다.

````python
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 4, 8]
plt.plot(x, y1, label='A 데이터')
plt.plot(x, y2, label='B 데이터')
plt.title('범례 위치 조정')
plt.legend(loc='upper left')
plt.show()
````

#### 2-4. 그리드(Grid) 추가하기

그래프 배경에 격자무늬를 추가하면 데이터 포인트의 정확한 값을 읽는 데 도움이 됩니다. `plt.grid(True)` 함수를 호출하여 그리드를 활성화할 수 있습니다. 그리드는 x, y축의 눈금에 맞춰 수평 및 수직선을 그어주어, 특정 지점의 좌표 값을 파악하기 용이하게 만듭니다. `axis` 파라미터('x' 또는 'y')를 사용해 특정 축의 그리드만 표시할 수도 있습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('그리드 추가')
plt.grid(True)
plt.show()
````

#### 2-5. 축 범위(Limits) 지정하기

Matplotlib는 기본적으로 데이터의 최소값과 최대값에 맞춰 자동으로 축의 범위를 설정합니다.

하지만 때로는 특정 부분을 확대해서 보거나, 여러 그래프의 축 범위를 동일하게 맞춰 비교하고 싶을 때가 있습니다.

`plt.xlim()`과 `plt.ylim()` 함수를 사용하여 각각 x축과 y축의 최소, 최대값을 직접 지정할 수 있습니다.

이를 통해 데이터의 특정 영역에 집중하거나 비교 기준을 통일할 수 있습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('축 범위 지정')
plt.xlim(0, 5) # X축 범위: 0~5
plt.ylim(0, 20) # Y축 범위: 0~20
plt.show()
````

#### 2-6. 텍스트 및 주석 추가하기

그래프에 추가적인 정보를 제공하기 위해 텍스트나 주석을 삽입할 수 있습니다. 

`plt.text()`는 지정된 좌표(x, y)에 텍스트를 추가하는 함수입니다. 

`plt.annotate()`는 한 단계 더 나아가, 특정 데이터 포인트(xy)를 가리키는 화살표와 함께 설명 텍스트(xytext)를 추가할 수 있는 강력한 기능을 제공합니다.

이를 통해 특정 데이터의 의미를 강조하거나 부가적인 설명을 덧붙여 그래프의 정보 전달력을 크게 향상시킬 수 있습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('텍스트와 주석')
# (2, 5) 위치에 텍스트 삽입
plt.text(2, 5, '여기는 (2, 5) 지점')
# (4, 16) 점에 주석 달기
plt.annotate('최대값!', xy=(4, 16), xytext=(2.5, 17), arrowprops=dict(facecolor='black', shrink=0.05))
plt.show()
````

#### 2-7. 수학적 표현 사용하기

Matplotlib는 TeX 마크업 문법을 지원하여, 그래프의 제목, 레이블, 주석 등에 분수, 제곱, 그리스 문자 등 다양한 수학적 표현식을 손쉽게 삽입할 수 있습니다. 

텍스트 문자열을 달러 기호(`$`)로 감싸고, 그 안에 TeX 문법에 따라 수식을 작성하면 됩니다. 

예를 들어, `$y = x^2$`와 같이 작성하면 일반 텍스트가 아닌 수학 식의 형태로 아름답게 렌더링됩니다. 

이를 통해 과학 및 공학 분야의 데이터를 시각화할 때 수식을 정확하고 전문적으로 표현할 수 있습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('수학적 표현 사용')
plt.xlabel('x축')
plt.ylabel('$y = x^2$', fontsize=14) # y축 레이블에 수식 적용
plt.text(2, 12, '$x^2 + y^2 = r^2$', fontsize=16) # 원의 방정식 삽입
plt.show()
````

### Part 3. 그래프 스타일 꾸미기

#### 3-1. 색상(Color) 지정하기

그래프 요소의 색상을 지정하는 것은 시각적 구분을 명확하게 하는 가장 기본적인 방법입니다.

`plot` 함수 내의 `color` 파라미터를 사용해 선의 색상을 변경할 수 있습니다. 

색상은 'red', 'blue'와 같은 이름으로 지정할 수도 있고, 'r', 'g', 'b'와 같은 약어로도 가능합니다.

또한, '#FF0000'과 같은 16진수 색상 코드를 사용하여 매우 정교한 색상 표현도 가능합니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y, color='red')
plt.title('색상 지정')
plt.show()
````

#### 3-2. 선 스타일(Line Style) 지정하기

여러 선이 겹쳐 있을 때, 색상만으로 구분하기 어려울 수 있습니다. 

이때 선의 스타일을 다르게 하면 구분이 더 쉬워집니다. `linestyle` 파라미터를 사용하며, 주요 값으로는 실선 `'-'` (기본값), 파선 `'--'`, 점선 `':'`, 파선과 점선이 섞인 `'-.'` 등이 있습니다. 

데이터의 성격에 따라(예: 예측 데이터는 파선으로) 선 스타일을 다르게 지정하면 정보 전달력을 높일 수 있습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y, linestyle='--')
plt.title('선 스타일 지정')
plt.show()
````

#### 3-3. 마커(Marker) 지정하기

마커는 데이터가 실제로 존재하는 위치(좌표)에 점을 찍어주는 기능입니다.

선 그래프에서 데이터 포인트의 정확한 위치를 강조하고 싶을 때 유용합니다. 

`marker` 파라미터를 사용하며, 원 `'o'`, 사각형 `'s'`, 별 `'*'`, 삼각형 `'^'`, 플러스 `'+'` 등 다양한 모양을 지정할 수 있습니다. 

마커를 사용하면 데이터의 밀도나 개별 데이터 포인트를 더 명확하게 인지할 수 있습니다.

#### 3-7. 폰트(Font) 관련 설정

#### 폰트 크기 지정하기

그래프의 가독성을 높이기 위해 제목, 축 레이블 등의 폰트 크기를 조절할 수 있습니다. `title()`, `xlabel()`, `ylabel()`과 같은 텍스트 관련 함수들에는 `fontsize` 파라미터가 있어, 원하는 크기를 숫자로 지정할 수 있습니다. 발표 자료에 사용될 그래프는 크게, 보고서에 사용될 그래프는 작게 만드는 등, 사용 목적에 맞게 폰트 크기를 조절하여 정보 전달의 효율성을 높일 수 있습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y)
plt.title('제목 폰트 크기 지정', fontsize=20)
plt.xlabel('X축', fontsize=14)
plt.ylabel('Y축', fontsize=14)
plt.show()
````

#### 폰트 딕셔너리 활용하기

폰트의 여러 속성(글꼴, 크기, 색상, 굵기 등)을 한 번에 관리하고 싶을 때 딕셔너리를 사용하면 편리합니다.

폰트 속성을 담은 딕셔너리를 만든 후, 텍스트 관련 함수의 `fontdict` 파라미터에 전달하면 해당 스타일이 일괄 적용됩니다. 

이를 통해 코드의 재사용성을 높이고, 그래프 전체의 폰트 스타일을 일관되게 유지하기 용이합니다. 

예를 들어, 제목용 폰트와 레이블용 폰트를 미리 정의해두고 사용할 수 있습니다.

````python
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
font1 = {'family': 'Malgun Gothic', 'size': 24, 'color':  'black'}
font2 = {'family': 'Malgun Gothic', 'size': 18, 'weight': 'bold', 'color':  'darkred'}

plt.plot(x, y)
plt.title('폰트 딕셔너리 활용', fontdict=font1)
plt.xlabel('X축', fontdict=font2)
plt.ylabel('Y축', fontdict=font2)
plt.show()
````

#### 한글 폰트 깨짐 문제 해결

Matplotlib의 기본 폰트는 한글을 지원하지 않아 글자가 깨져 네모(□)로 보일 수 있습니다.

이를 해결하기 위해 한글을 지원하는 폰트로 설정해야 합니다.

**해결 방법 1: 전체 폰트 설정 (권장)**

노트북 시작 부분에서 `plt.rc()`를 사용해 전체 폰트를 설정하는 방법입니다.
`rc`는 'runtime configuration'의 약자로, Matplotlib의 실행 환경 설정을 변경하는 데 사용됩니다. 

`font`의 `family`를 'Malgun Gothic'(Windows)이나 'AppleGothic'(macOS) 등 시스템에 설치된 한글 폰트로 지정하면, 이후 모든 그래프에 해당 폰트가 적용되어 편리합니다.

````python
# Windows: 'Malgun Gothic'
# macOS: 'AppleGothic'
plt.rc('font', family='Malgun Gothic')
plt.rc('axes', unicode_minus=False) # 마이너스 기호 깨짐 방지

plt.title('한글 제목')
plt.plot([-1, 0, 1], [-1, 0, 1])
plt.show()
````

**해결 방법 2: 시스템에 설치된 폰트 확인하기**

사용하고 싶은 다른 폰트가 있거나, 현재 환경에 어떤 폰트가 설치되어 있는지 확인하고 싶을 때 `matplotlib.font_manager`를 사용합니다. 

`fm.findSystemFonts()`는 시스템의 모든 폰트 파일 경로를 찾아 리스트로 반환합니다.

이 경로를 `fm.FontProperties`에 전달하면 해당 폰트의 이름을 얻을 수 있습니다. 

이를 통해 사용 가능한 한글 폰트를 확인하고 `plt.rc()`에 설정할 수 있습니다.

````python
import matplotlib.font_manager as fm

# 모든 폰트 목록 가져오기
font_list = fm.findSystemFonts(fontpaths=None, fontext='ttf')

# 'Malgun' 또는 'Gothic'이 포함된 폰트만 출력
for font_path in font_list:
    font_name = fm.FontProperties(fname=font_path).get_name()
    if 'malgun' in font_name.lower() or 'gothic' in font_name.lower():
        print(font_name, ':', font_path)
````

### Part 4. 다양한 그래프 그리기

#### 4-1. 막대 그래프 (Bar Chart)

막대 그래프는 범주형 데이터의 값(크기, 양, 빈도 등)을 직사각형 막대의 길이로 표현하는 시각화 방법입니다. 

각 막대는 독립된 항목을 나타내므로, 여러 항목 간의 값을 직관적으로 비교하는 데 매우 효과적입니다. 

`plt.bar()` 함수를 사용하며, 첫 번째 인자로는 범주(x축), 두 번째 인자로는 각 범주의 값(y축)을 전달합니다. 

제품별 판매량, 국가별 인구수 등 명확한 구분이 있는 데이터에 적합합니다.

````python
products = ['A', 'B', 'C', 'D']
quantities = [15, 30, 22, 40]
plt.bar(products, quantities)
plt.title('제품별 판매량')
plt.show()
````

#### 막대 그래프에 패턴 채우기

흑백으로 인쇄하거나 색상 구분이 어려운 환경을 위해 막대 내부에 패턴을 채울 수 있습니다.

`hatch` 파라미터를 사용하면 됩니다. 

`'/'`, `'\'`, `'|'`, `'-'`, `'+'`, `'x'`, `'o'`, `'O'`, `'.'`, `'*'` 등 다양한 패턴을 지정하여 각 막대를 시각적으로 구분할 수 있습니다.

색상과 패턴을 함께 사용하면 더 풍부하고 명확한 정보 전달이 가능합니다.

#### 4-2. 가로 막대 그래프 (Horizontal Bar Chart)

가로 막대 그래프는 세로 막대 그래프를 90도 회전시킨 형태로, y축에 범주를, x축에 값을 표시합니다.

순위나 순서를 나타낼 때 위에서부터 아래로 시선이 자연스럽게 이동하여 더 효과적으로 정보를 전달하기도 합니다.

#### 4-3. 산점도 (Scatter Plot)

산점도는 두 개의 연속형 변수 간의 관계를 파악하기 위해 점으로 데이터를 나타내는 그래프입니다.

각 점은 하나의 관측치를 의미하며, x축과 y축의 값에 따라 좌표 평면에 위치합니다.

`plt.scatter()` 함수를 사용합니다.

점들이 모여있는 패턴을 통해 두 변수 간의 상관관계(양의 상관, 음의 상관, 무상관)나 데이터의 군집, 이상치 등을 시각적으로 파악할 수 있습니다. 

키와 몸무게, 공부 시간과 성적 등의 관계를 분석하는 데 널리 사용됩니다.

````python
study_time = [2, 3, 5, 1, 6, 4, 7, 2, 8, 5]
scores = [65, 70, 85, 60, 90, 75, 95, 68, 98, 82]
plt.scatter(study_time, scores)
plt.title('공부 시간과 성적')
plt.show()
````

#### 산점도에 컬러맵 적용하기

산점도에 세 번째 변수를 추가하여 더 풍부한 정보를 표현할 수 있습니다. 

`c` 파라미터에 세 번째 변수(수치형)를 전달하고, `cmap` 파라미터에 컬러맵의 이름을 지정하면, 해당 변수의 값에 따라 점의 색상이 연속적으로 변하게 됩니다. 

`plt.colorbar()` 함수를 호출하면 색상 막대가 추가되어 색상과 실제 값의 관계를 보여줍니다.

'viridis', 'plasma', 'coolwarm' 등 다양한 내장 컬러맵을 사용할 수 있습니다.

````python
study_time = [2, 3, 5, 1, 6, 4, 7, 2, 8, 5]
scores = [65, 70, 85, 60, 90, 75, 95, 68, 98, 82]
satisfaction = [3, 4, 5, 2, 5, 4, 5, 3, 5, 4] # 세 번째 변수 (만족도)
plt.scatter(study_time, scores, c=satisfaction, cmap='viridis')
plt.colorbar(label='수업 만족도')
plt.title('공부 시간, 성적, 만족도의 관계')
plt.show()
````

#### 4-4. 히스토그램 (Histogram)

히스토그램은 하나의 연속형 변수에 대한 데이터의 분포를 시각화하는 데 사용됩니다.

전체 데이터 범위를 여러 개의 동일한 간격(구간, `bins`)으로 나눈 후, 각 구간에 속하는 데이터의 개수(빈도)를 막대의 높이로 표현합니다.

`plt.hist()` 함수를 사용합니다.

데이터가 어떤 값에 집중되어 있는지, 얼마나 넓게 퍼져 있는지, 분포가 대칭적인지 등을 한눈에 파악할 수 있습니다.

시험 점수 분포나 소득 분포 등을 분석할 때 유용합니다.

````python
math_scores = np.random.normal(loc=70, scale=15, size=100)
plt.hist(math_scores, bins=10) # bins: 구간의 개수
plt.title('수학 점수 분포')
plt.show()
````

#### 4-5. 파이 차트 (Pie Chart)

파이 차트는 전체에 대한 각 부분의 비율을 원형의 부채꼴로 나누어 표현하는 그래프입니다.

각 부채꼴의 면적이나 중심각이 해당 항목의 비율을 나타냅니다.

`plt.pie()` 함수를 사용하며, `labels`로 각 항목의 이름을, `autopct`로 각 항목의 비율을 퍼센트로 표시할 수 있습니다. 

시장 점유율이나 설문조사 응답 비율처럼, 전체 합이 100%가 되는 데이터를 시각화할 때 주로 사용됩니다.

다만, 항목이 너무 많거나 비율이 비슷하면 비교가 어려워지는 단점이 있습니다.

````python
fruits = ['사과', '바나나', '딸기', '오렌지']
ratio = [30, 25, 25, 20]
plt.pie(ratio, labels=fruits, autopct='%.1f%%')
plt.title('선호 과일 비율')
plt.show()
````

#### 4-6. 박스 플롯 (Box Plot)

박스 플롯은 수치형 데이터의 분포를 사분위수를 이용하여 시각화하는 방법입니다. 

`plt.boxplot()` 함수를 사용하며, 데이터의 중앙값(median), 1사분위수(Q1), 3사분위수(Q3), 그리고 최소/최대값과 이상치(outlier)를 한눈에 보여줍니다. 

데이터의 대칭성, 퍼진 정도, 이상치의 유무 등을 빠르게 파악할 수 있습니다. 

여러 그룹의 데이터 분포를 나란히 놓고 비교할 때 매우 효과적입니다.

Seaborn을 사용하면 더 미려하고 간편하게 그릴 수 있습니다.

````python
data_a = np.random.normal(0, 2.0, 100)
data_b = np.random.normal(-3.0, 1.5, 100)
plt.boxplot([data_a, data_b], labels=['A 데이터', 'B 데이터'])
plt.title('Matplotlib 박스 플롯')
plt.show()
````

#### 4-7. 바이올린 플롯 (Violin Plot)

바이올린 플롯은 박스 플롯과 데이터의 밀도를 나타내는 커널 밀도 추정(KDE)을 결합한 형태입니다. 

`plt.violinplot()` 함수를 사용합니다. 박스 플롯처럼 데이터의 요약 통계량을 보여주면서도, 데이터가 어떤 값에 많이 분포하는지(밀도)를 바이올린 모양으로 함께 보여줍니다. 

데이터의 분포 형태(예: 봉우리가 하나인 단봉분포, 여러 개인 다봉분포)를 더 상세하게 파악할 수 있는 장점이 있습니다.

````python
data_a = np.random.normal(0, 2.0, 100)
data_b = np.random.normal(-3.0, 1.5, 100)
plt.violinplot([data_a, data_b])
plt.xticks([1, 2], ['A 데이터', 'B 데이터'])
plt.title('Matplotlib 바이올린 플롯')
plt.show()
````

### Part 5. 객체지향 API 활용

지금까지 사용한 `plt.plot()`, `plt.title()` 같은 함수들은 Matplotlib이 내부적으로 Figure와 Axes 객체를 만들고 관리해주는 방식입니다. 

더 정교한 제어를 위해 이 객체들을 직접 다루는 방법을 알아봅니다. 

객체지향 API를 사용하면 하나의 Figure 안에 여러 개의 그래프(Axes)를 배치하거나 각 요소들을 개별적으로 세밀하게 제어할 수 있어 복잡한 시각화를 구현하는 데 필수적입니다.

#### 5-1. Figure와 Axes 객체 생성하기

`plt.subplots()` 함수는 Figure 객체와 Axes 객체를 쌍으로 반환합니다. 

Figure(`fig`)는 전체 그림이 그려질 도화지나 캔버스에 해당하며, 그래프의 크기나 전체적인 배경을 관리합니다. 

Axes(`ax`)는 실제로 그래프가 그려지는 개별 공간으로, x축, y축, 플롯, 제목 등 대부분의 그래프 요소를 포함하고 있습니다.

이 두 객체를 변수로 받아 사용함으로써 그래프의 모든 요소를 명시적으로 제어할 수 있게 됩니다.

````python
# fig는 전체 도화지, ax는 그림을 그릴 공간입니다.
fig, ax = plt.subplots()
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
ax.plot(x, y)
plt.show()
````

#### 5-2. 객체지향 방식으로 요소 설정하기

객체지향 방식에서는 `plt.title()` 대신 `ax.set_title()`을, `plt.xlabel()` 대신 `ax.set_xlabel()`을 사용하는 식으로 함수 이름이 약간 다릅니다. 

대부분의 경우 `plt`로 시작하던 함수 이름이 `ax.set_*` 형태로 바뀐다고 생각하면 쉽습니다. 

(예: `ax.set_ylabel()`, `ax.set_xlim()`) 이처럼 Axes 객체(`ax`)의 메서드를 직접 호출하여 그래프의 각 요소를 설정하는 것이 객체지향 API의 핵심입니다.

````python
fig, ax = plt.subplots()
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
ax.plot(x, y, label='데이터')
ax.set_title('객체지향 방식 그래프') # set_title()
ax.set_xlabel('X축') # set_xlabel()
ax.set_ylabel('Y축') # set_ylabel()
ax.legend() # legend()
ax.grid(True) # grid()
plt.show()
````

#### 5-3. 여러 그래프 함께 그리기 (Subplots)

객체지향 API의 가장 큰 장점 중 하나는 여러 개의 그래프를 한 번에 그리는 것입니다. 

`plt.subplots(nrows, ncols)` 함수에 행과 열의 개수를 지정하면, 해당 개수만큼의 Axes 객체를 담은 배열을 반환합니다.

예를 들어, `plt.subplots(2, 2)`는 2x2 형태의 4개 그래프 공간을 생성합니다.

반환된 `axes` 배열의 각 요소에 인덱스로 접근(`axes[0, 0]`, `axes[0, 1]`)하여 각각의 그래프를 독립적으로 제어할 수 있습니다.

````python
# 1행 2열의 서브플롯을 생성합니다.
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

# 첫 번째 그래프 (axes[0])
axes[0].plot(x, y, 'ro-')
axes[0].set_title('첫 번째 그래프')

# 두 번째 그래프 (axes[1])
axes[1].plot(y, x, 'bs--')
axes[1].set_title('두 번째 그래프')

fig.suptitle('서브플롯 예제', fontsize=16) # 전체 제목
plt.tight_layout() # 그래프 간격 자동 조절
plt.show()
````

#### 5-4. 한 공간에 다른 종류 그래프 그리기

하나의 Axes 객체에 `plot()`, `bar()` 등 여러 종류의 그리기 함수를 연속해서 사용하면, 다른 종류의 그래프를 한 공간에 겹쳐서 그릴 수 있습니다.

예를 들어, 막대 그래프 위에 선 그래프를 함께 그려서 전체적인 추세와 각 항목의 값을 동시에 보여줄 수 있습니다.

이 경우 각 그래프의 데이터와 축의 의미가 명확히 전달되도록 범례와 축 레이블을 정확히 설정하는 것이 매우 중요합니다.

````python
fig, ax = plt.subplots()
products = ['A', 'B', 'C', 'D']
sales = [15, 30, 22, 40]
growth_rate = [0.2, 0.5, 0.1, 0.3] # 성장률
predicted_sales = np.array(sales) * (1 + np.array(growth_rate))

ax.bar(products, sales, label='현재 판매량')
ax.plot(products, predicted_sales, 'ro-', label='성장 후 예측')

ax.set_title('제품별 판매량과 성장 예측')
ax.legend()
plt.show()
````

### Part 6. Matplotlib 고급 기능

#### 6-1. 축 위치 및 스타일 변경 (Spines)

그래프의 테두리를 이루는 선을 `Spine`이라고 합니다. 기본적으로 상하좌우에 모두 표시되지만, 이를 제어하여 더 깔끔하거나 독특한 스타일의 그래프를 만들 수 있습니다. 

Axes 객체의 `spines` 속성을 통해 각 위치(top, bottom, left, right)의 Spine에 접근할 수 있습니다.

`set_visible(False)`로 특정 축을 보이지 않게 하거나, `set_position()`으로 축의 위치를 데이터 영역 안으로 옮기는 등(예: 0을 기준으로 교차하는 좌표축) 다양한 연출이 가능합니다.

````python
fig, ax = plt.subplots()
x = np.linspace(-np.pi, np.pi, 256)
y = np.cos(x)
ax.plot(x, y)
# 위쪽, 오른쪽 축 숨기기
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
# 아래쪽 축을 y=0 위치로 이동
ax.spines['bottom'].set_position(('data', 0))
# 왼쪽 축을 x=0 위치로 이동
ax.spines['left'].set_position(('data', 0))
plt.title('축 위치/스타일 변경')
plt.show()
````

#### 6-2. 이중 Y축 사용하기 (twinx)

서로 다른 단위나 스케일을 가진 두 종류의 데이터를 하나의 그래프에 함께 표시하고 싶을 때 이중 Y축을 사용합니다.

Axes 객체의 `twinx()` 메서드는 기존의 x축을 공유하면서 새로운 y축을 가지는 별개의 Axes 객체를 생성합니다.

이를 통해 왼쪽 y축과 오른쪽 y축이 각각 다른 데이터 계열을 설명하게 할 수 있습니다.

예를 들어, 월별 강수량(mm)과 평균 기온(°C)을 한 번에 비교 분석할 수 있습니다.

````python
fig, ax1 = plt.subplots()
x = np.arange(1, 13)
y1 = np.random.randint(50, 200, 12)   # 월별 강수량
y2 = np.random.uniform(10, 25, 12)    # 월별 평균 기온

color = 'tab:blue'
ax1.set_xlabel('월')
ax1.set_ylabel('강수량 (mm)', color=color)
ax1.bar(x, y1, color=color, alpha=0.6)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()  # x축을 공유하는 두 번째 y축
color = 'tab:red'
ax2.set_ylabel('평균 기온 (°C)', color=color)
ax2.plot(x, y2, color=color, marker='o')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('월별 강수량과 평균 기온')
fig.tight_layout()
plt.show()
````

#### 6-3. 도형 삽입하기 (Patches)

그래프에 사각형, 원, 다각형 등 다양한 도형을 추가하여 특정 영역을 강조하거나 부가적인 시각 요소를 더할 수 있습니다. 

`matplotlib.patches` 모듈에 있는 `Rectangle`, `Circle`, `Polygon` 등의 클래스를 사용합니다.

도형 객체를 생성한 후, Axes 객체의 `add_patch()` 메서드를 사용하여 그래프에 추가합니다.

이를 통해 데이터의 특정 구간을 강조하거나, 다이어그램과 같은 복합적인 시각 자료를 만들 수 있습니다.

````python
import matplotlib.patches as patches

fig, ax = plt.subplots()
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
ax.plot(x, y)
ax.set_xlim(0, 5)
ax.set_ylim(0, 17)

# 사각형 추가: 시작점(1, 2), 너비 2, 높이 5
rect = patches.Rectangle((1, 2), 2, 5, linewidth=1, edgecolor='r', facecolor='none')
ax.add_patch(rect)

# 원 추가: 중심(3, 12), 반지름 2
circ = patches.Circle((3, 12), 2, linewidth=1, edgecolor='b', facecolor='skyblue', alpha=0.5)
ax.add_patch(circ)

plt.title('도형 삽입하기')
ax.set_aspect('equal', adjustable='box') # 도형의 비율을 유지하기 위해 추가
plt.show()
````

#### 6-4. 그래프 이미지로 저장하기

완성된 그래프는 보고서나 발표 자료 등에 활용하기 위해 이미지 파일로 저장해야 할 때가 많습니다. 

`plt.savefig()` 함수를 사용하면 현재 Figure에 그려진 그래프를 파일로 저장할 수 있습니다.

파일 이름을 인자로 전달하며, 확장자에 따라 PNG, JPG, PDF, SVG 등 다양한 포맷으로 저장이 가능합니다.

`dpi` 파라미터로 해상도(dots per inch)를 조절할 수 있으며, `bbox_inches='tight'` 옵션은 그래프 주변의 불필요한 여백을 제거해줍니다.

````python
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 4, 8]
plt.plot(x, y1, label='A 데이터')
plt.plot(x, y2, label='B 데이터')
plt.title('저장할 그래프')
plt.legend()

# 'my_plot.png' 라는 이름으로 현재 디렉토리에 저장
# dpi는 해상도를 의미하며, 숫자가 클수록 고화질
plt.savefig('my_plot.png', dpi=300, bbox_inches='tight')

plt.show()
````

#### 6-2. 카운트 플롯 (Count Plot)

카운트 플롯은 범주형 변수에 속하는 데이터의 개수(빈도)를 세어 막대 그래프로 보여주는 함수입니다. Matplotlib의 `bar` 함수와 달리, y축의 값을 직접 계산할 필요 없이 x축(또는 y축)에 범주형 변수만 지정해주면 Seaborn이 알아서 각 범주의 개수를 세어 그려줍니다. 특정 항목이 데이터셋에 얼마나 많이 나타나는지 확인할 때 매우 유용합니다.

````python
# 요일(day)별 데이터의 개수를 세어 막대 그래프로 보여줍니다.
sns.countplot(data=tips, x='day')
plt.title('요일별 방문 횟수')
plt.show()
````

#### 6-3. 박스 플롯 (Box Plot)

박스 플롯은 수치형 데이터의 분포를 사분위수를 이용하여 시각화하는 방법입니다. 데이터의 중앙값(50%), 25% 지점, 75% 지점, 그리고 최소/최대값과 이상치(outlier)를 한눈에 보여줍니다. 이를 통해 데이터의 대칭성, 퍼진 정도, 이상치의 유무 등을 빠르게 파악할 수 있습니다. `sns.boxplot()` 함수를 사용하며, 범주별로 데이터 분포를 비교할 때 매우 효과적입니다.

````python
# 요일(day)별 팁(tip)의 분포를 보여줍니다.
# 박스는 데이터의 25%, 50%(중앙값), 75% 지점을 나타냅니다.
sns.boxplot(data=tips, x='day', y='tip')
plt.title('요일별 팁 분포')
plt.show()
````

#### 6-4. 바이올린 플롯 (Violin Plot)

바이올린 플롯은 박스 플롯과 커널 밀도 추정(KDE) 그래프를 합쳐놓은 형태입니다. 데이터의 사분위수 정보를 보여주는 박스 플롯의 특징과 함께, 데이터가 어떤 값에 많이 분포하는지를 나타내는 밀도 곡선을 함께 보여줍니다. `sns.violinplot()`을 사용하며, 데이터의 분포 형태(예: 봉우리가 하나인 단봉분포, 여러 개인 다봉분포)를 더 상세하게 파악할 수 있는 장점이 있습니다.

````python
# 박스 플롯과 데이터의 밀도(분포)를 함께 보여줍니다.
sns.violinplot(data=tips, x='day', y='tip')
plt.title('요일별 팁 분포 (밀도 포함)')
plt.show()
````

#### 6-5. 산점도 (Scatter Plot) with hue

Seaborn의 산점도는 Matplotlib보다 더 많은 정보를 담을 수 있습니다. 특히 `hue` 파라미터는 세 번째 범주형 변수를 기준으로 점의 색상을 다르게 하여, 두 변수의 관계를 그룹별로 나누어 볼 수 있게 해줍니다. 예를 들어, 총 식사 금액과 팁의 관계를 흡연 여부(`smoker`)에 따라 다른 색으로 표시하면, 각 그룹의 패턴 차이를 명확하게 비교할 수 있습니다. `size`, `style` 파라미터를 이용해 더 많은 변수를 표현하는 것도 가능합니다.

````python
# total_bill과 tip의 관계를 점으로 나타냅니다.
# hue='smoker' 옵션으로 흡연 여부에 따라 점의 색을 다르게 표시합니다.
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='smoker')
plt.title('총 식사 금액과 팁의 관계 (흡연 여부별)')
plt.show()
````

#### 6-6. 히트맵 (Heatmap)

히트맵은 2차원 데이터를 색상으로 표현하는 시각화 기법입니다. 주로 행과 열을 가지는 매트릭스 형태의 데이터를 시각화하는 데 사용되며, 각 셀의 숫자 값을 색상의 농도나 종류로 변환하여 표현합니다. `sns.heatmap()`을 사용하며, `annot=True`로 설정하면 각 셀에 숫자 값을 표시할 수 있습니다. 변수 간의 상관관계 행렬이나, 시간-범주에 따른 데이터 변화(예: 월별/연도별 승객 수)를 한눈에 파악하는 데 매우 효과적입니다.

````python
# flights 데이터셋 로드
flights = sns.load_dataset('flights')
# 데이터를 연도-월 매트릭스로 변환
flights_pivot = flights.pivot(index='month', columns='year', values='passengers')

plt.figure(figsize=(10, 8))
sns.heatmap(flights_pivot, annot=True, fmt='d', cmap='YlGnBu')
plt.title('연도/월별 승객 수')
plt.show()
````

#### 6-7. 페어 플롯 (Pair Plot)

### plotly 라이브러리 사용

Plotly는 사용자의 상호작용에 반응하는 동적인 '인터랙티브(interactive)' 그래프를 만들 수 있는 강력한 시각화 라이브러리입니다. 

Matplotlib이나 Seaborn이 정적인 이미지를 생성하는 것과 달리, Plotly로 만든 그래프는 마우스를 올리면 데이터 값을 보여주거나, 

특정 영역을 확대/축소하고, 범례를 클릭하여 특정 데이터를 숨기거나 나타내는 등 다양한 상호작용이 가능합니다.

특히 웹 기반 대시보드나 데이터 분석 보고서에 매우 유용합니다.

`plotly.express`는 Plotly를 더 쉽고 간결하게 사용할 수 있도록 만든 고수준 API로, 적은 양의 코드로 매우 복잡하고 아름다운 그래프를 생성할 수 있습니다.

이 파트에서는 `plotly.express`를 중심으로 실습을 진행합니다.

#### 8-1. Plotly 소개 및 환경 설정

Plotly를 사용하기 위해서는 먼저 라이브러리를 설치해야 합니다. (`pip install plotly`) 그 후, 관례적으로 `px`라는 별칭으로 `plotly.express`를 임포트하여 사용합니다. Plotly는 웹 기반으로 동작하므로, 주피터 노트북이나 Colab 환경에서 별도의 설정 없이 바로 인터랙티브 그래프를 확인할 수 있습니다.

````python
import plotly.express as px
import pandas as pd
import numpy as np
import plotly.io as pio
pio.renderers.default = "notebook"

# Plotly에서 사용할 예제 데이터프레임 생성
df = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y1': [1, 4, 9, 16, 25],
    'y2': [5, 8, 6, 10, 7],
    'category': ['A', 'A', 'B', 'A', 'B']
})
````

#### 8-2. 라인 차트 (Line Chart)

`px.line()` 함수는 선 그래프를 생성합니다. 데이터프레임을 첫 번째 인자로 전달하고, `x`와 `y` 파라미터에 x축과 y축으로 사용할 컬럼 이름을 지정합니다. Matplotlib과 마찬가지로 데이터의 추세나 변화를 시각화하는 데 사용되지만, 마우스를 올리면 각 데이터 포인트의 정확한 값을 바로 확인할 수 있다는 장점이 있습니다.

````python
fig = px.line(df, x='x', y='y1', title='Plotly 기본 라인 차트')
fig.show()
````

■ 라인 차트 스타일링

Plotly에서는 `px` 함수 내에서 `title`, `labels`, `color_discrete_sequence` 등의 파라미터를 통해 기본적인 스타일을 지정할 수 있습니다. 더 세밀한 조정은 `update_layout()`과 `update_traces()` 메서드를 사용합니다. 예를 들어, `update_traces()`로 마커를 추가하고, `update_layout()`으로 제목 폰트와 축 제목을 변경할 수 있습니다.

````python
fig = px.line(df, x='x', y=['y1', 'y2'], 
              title='Plotly 스타일링 라인 차트',
              labels={'x': 'X축', 'value': 'Y축 값', 'variable': '데이터 종류'},
              color_discrete_sequence=['#FF5733', '#335BFF'] # 선 색상 지정
             )

# 마커 추가 및 선 스타일 변경
fig.update_traces(marker=dict(size=10), selector=dict(name='y1'))
fig.update_traces(line=dict(dash='dash'), selector=dict(name='y2'))

# 전체적인 레이아웃 업데이트
fig.update_layout(
    title_font_size=24,
    legend_title_text='데이터'
)

fig.show()
````

#### 8-3. 막대 차트 (Bar Chart)

`px.bar()` 함수는 막대 그래프를 생성합니다. 범주형 데이터 간의 값을 비교하는 데 사용됩니다. `color` 파라미터에 다른 컬럼을 지정하면 해당 컬럼의 값에 따라 막대의 색을 다르게 칠할 수 있어, 다차원적인 정보를 한 번에 표현하기 용이합니다. `text_auto=True` 옵션은 각 막대 위에 값을 자동으로 표시해주는 편리한 기능입니다.

````python
products = ['A', 'B', 'C', 'D']
quantities = [15, 30, 22, 40]
category = ['가', '나', '가', '나']

fig = px.bar(x=products, y=quantities, color=category,
             title='제품별 판매량 (카테고리별)',
             labels={'x': '제품', 'y': '판매량', 'color': '카테고리'},
             text_auto=True, # 막대 위에 값 표시
             template='plotly_white' # 깔끔한 템플릿 적용
            )
fig.show()
````

#### 8-4. 산점도 (Scatter Plot)

`px.scatter()` 함수는 두 연속형 변수 간의 관계를 점으로 나타냅니다. Plotly 산점도의 가장 큰 장점은 `color`, `size`, `symbol` 파라미터를 이용해 3차원, 4차원, 5차원의 정보를 하나의 2D 그래프에 표현할 수 있다는 점입니다. 또한 `hover_data`를 통해 마우스를 올렸을 때 나타나는 정보(툴팁)를 추가로 지정할 수 있습니다.

````python
# iris 데이터셋 로드
iris = px.data.iris()

fig = px.scatter(iris, x='sepal_width', y='sepal_length', 
                 color='species', # 종에 따라 색상 구분
                 size='petal_length', # 꽃잎 길이에 따라 점 크기 조절
                 hover_data=['petal_width'], # 툴팁에 꽃잎 너비 정보 추가
                 title='Iris 데이터 산점도',
                 labels={'sepal_width': '꽃받침 너비', 'sepal_length': '꽃받침 길이', 'species': '종'}
                )
fig.show()
````

#### 8-5. 파이 차트 (Pie Chart)

`px.pie()` 함수는 전체에 대한 각 부분의 비율을 부채꼴로 나타냅니다. `names` 파라미터에는 항목의 이름을, `values` 파라미터에는 각 항목의 값을 전달합니다. `hole` 파라미터에 0과 1 사이의 값을 주면 중앙이 비어있는 도넛 차트를 만들 수 있어, 더 세련된 시각화를 구현할 수 있습니다.

````python
fruits = ['사과', '바나나', '딸기', '오렌지']
ratio = [30, 25, 25, 20]

fig = px.pie(names=fruits, values=ratio, 
             title='선호 과일 비율',
             hole=0.3, # 도넛 차트 만들기
             color_discrete_sequence=px.colors.sequential.RdBu # 색상 팔레트 적용
            )

# 텍스트 정보와 퍼센트를 함께 표시
fig.update_traces(textinfo='percent+label')
fig.show()
````

#### 8-6. 히스토그램 (Histogram)

`px.histogram()` 함수는 연속형 변수의 분포를 막대 형태로 보여줍니다. `nbins` 파라미터로 막대의 개수를 조절할 수 있으며, `marginal` 파라미터에 'rug', 'box', 'violin' 등을 지정하면 그래프의 상단이나 오른쪽에 해당 변수의 분포를 다른 형태로 추가하여 보여주는 기능(marginal plot)을 사용할 수 있습니다.

````python
tips = px.data.tips()

fig = px.histogram(tips, x='total_bill',
                  title='총 식사 금액 분포',
                  nbins=30, # 막대 개수
                  marginal='rug', # 위쪽에 러그 플롯 추가
                  color='sex' # 성별에 따라 색상 구분
                 )
fig.show()
````

#### 8-7. 박스 플롯 (Box Plot)

`px.box()` 함수는 데이터의 사분위수 분포를 시각화합니다. Seaborn과 마찬가지로 범주형 변수에 따른 수치형 변수의 분포를 비교하는 데 매우 효과적입니다. `points` 파라미터를 'all'로 설정하면 모든 데이터 포인트를 함께 표시하여 데이터의 실제 분포를 더 자세히 볼 수 있습니다.

````python
tips = px.data.tips()

fig = px.box(tips, x='day', y='tip',
             title='요일별 팁 분포',
             color='smoker', # 흡연 여부에 따라 박스 구분
             points='all' # 모든 데이터 포인트 표시
            )
fig.show()
````

````python
import folium
from folium.features import CustomIcon, DivIcon, IFrame

# 1) 지도 생성
m = folium.Map(location=[37.5665, 126.9780], zoom_start=13)

# 2) 커스텀 아이콘 이미지 (PNG, JPG 가능)
custom_icon = CustomIcon(
    icon_image="https://cdn-icons-png.flaticon.com/512/684/684908.png",  # 원하는 이미지 URL 또는 로컬 파일 경로
    icon_size=(50, 50),   # 이미지 크기(px)
    icon_anchor=(25, 50), # 기준점 (이미지 하단 중앙을 기준점으로)
)

# 3) 팝업 부분에 들어갈 HTML 정의 (자유롭게 꾸밀 수 있음)
html = """
<div style="font-family:Arial; width:500px;">
    <h4 style='margin-bottom:5px; color:#2A81CB;'>📍 Custom Marker</h4>
    <p style='margin:0;'>이 마커는 커스텀 아이콘을 사용합니다.</p>
    <p style='margin:0;'>원하는 HTML을 모두 넣을 수 있습니다.</p>
    <hr>
    <img src='https://picsum.photos/250/120' style='border-radius:8px;'>
    <p style='color:gray; font-size:12px;'>임의의 이미지도 삽입 가능</p>
</div>
"""

iframe = IFrame(html, width=260, height=250)
popup = folium.Popup(iframe, max_width=300)

# 4) Marker 생성
folium.Marker(
    location=[37.5665, 126.9780], 
    tooltip="✔ 커스텀 툴팁 예시", 
    popup=popup, 
    icon=custom_icon
).add_to(m)

m
````

단순 표시/마커 용도 (지도 확대/축소해도 화면 상 크기 거의 고정)

👉 공통/주요 옵션 정리

- location=[lat, lon] : 중심 좌표
- radius

    Circle: 미터

    CircleMarker: 픽셀 크기

- color : 테두리 선 색상
- weight : 선 두께(px)
- opacity : 선 투명도(0~1)
- fill : 내부를 채울지 여부(True/False)
- fill_color : 내부 채우기 색
- fill_opacity : 내부 투명도(0~1)
- popup : 클릭 시 뜨는 팝업
- tooltip : 마우스 오버 시 뜨는 텍스트
- dash_array : 테두리를 점선으로 (예: "5, 5")

````python
import folium

m = folium.Map(location=[37.5665, 126.9780], zoom_start=13)

# 1) 반경 1km를 가지는 Circle (실제 거리 표현)
folium.Circle(
    location=[37.5665, 126.9780],   # 중심 (서울시청 근처)
    radius=1000,                    # 미터 단위 (1km)
    color="blue",                   # 테두리 색
    weight=2,                       # 테두리 두께
    opacity=0.8,                    # 테두리 투명도
    fill=True,                      # 내부 채움 여부
    fill_color="blue",              # 내부 색
    fill_opacity=0.2,               # 내부 투명도
    popup=folium.Popup("반경 1km 영역", max_width=200),
    tooltip="서울시청 반경 1km",
).add_to(m)

# 2) 픽셀 단위 크기 CircleMarker (zoom해도 화면상 크기 비슷)
folium.CircleMarker(
    location=[37.5700, 126.9820],
    radius=10,                      # 픽셀 단위
    color="red",
    weight=3,
    opacity=1.0,
    fill=True,
    fill_color="#FF7777",
    fill_opacity=0.9,
    popup="픽셀 크기 Marker",
    tooltip="CircleMarker 예시",
).add_to(m)

m
````

👉 PolyLine 주요 옵션

- color : 선 색상
- weight : 선 두께

- line_cap : 끝 모양 ("round", "square", "butt")

````python
#폴리라인 예시
import folium

m = folium.Map(location=[37.5665, 126.9780], zoom_start=15)

path = [
    [37.5665, 126.9780],  # 서울시청
    [37.5700, 126.9820],  # 광화문 쪽
    [37.5750, 126.9768],  # 경복궁 쪽
]

folium.PolyLine(
    locations=path,
    color="red",
    weight=5,
    opacity=0.8,
    dash_array="10,5",      # 10px 선, 5px 공백 반복
    line_cap="round",
    line_join="round",
    popup="이동 경로 예시",
    tooltip="PolyLine 경로",
).add_to(m)

m
````

##### ✔️ GeoJSON Layer

✅ 개념

행정구역/국가 경계/서비스 커버리지 등의 복잡한 도형 데이터를
GeoJSON 형식으로 지도에 올릴 수 있음.

속성값에 따라 색 지정 → Choropleth 시각화 가능.

👉 folium.GeoJson 주요 옵션

- data : GeoJSON 데이터 (dict, 파일 내용, 경로 등)
- name : 레이어 이름 (LayerControl에서 표시)
- style_function

    각 feature별 스타일 지정 함수

    인자: feature (GeoJSON feature dict)

    반환: 스타일 dict ({"fillColor": ..., "color": ..., ...})
- highlight_function : 마우스 오버 시 스타일 변경
- tooltip : folium.GeoJsonTooltip
- popup : folium.GeoJsonPopup
- smooth_factor : 도형 부드럽게 처리(렌더링 최적화)
- show : 기본 표시 여부

````python
import folium
import json

m = folium.Map(location=[37.5665, 126.9780], zoom_start=12)

# 예시용 간단 GeoJSON (사각형 하나짜리 Feature)
geojson_data = {
    "type": "FeatureCollection",
    "features": [
        {
            "type": "Feature",
            "properties": {"name": "Example Area", "value": 99},
            "geometry": {
                "type": "Polygon",
                "coordinates": [[
                    [126.97, 37.57],
                    [126.99, 37.57],
                    [126.99, 37.56],
                    [126.97, 37.56],
                    [126.97, 37.57],
                ]]
            }
        }
    ]
}

def style_function(feature):
    # 속성에 따라 색 지정 가능
    val = feature["properties"]["value"]
    color = "green" if val < 100 else "red"
    return {
        "fillColor": color,
        "color": "black",
        "weight": 2,
        "fillOpacity": 0.4,
    }

def highlight_function(feature):
    return {
        "fillColor": "#ffff00",
        "color": "black",
        "weight": 3,
        "fillOpacity": 0.7,
    }

geo = folium.GeoJson(
    data=geojson_data,
    name="예시 GeoJSON",
    style_function=style_function,
    highlight_function=highlight_function,
    tooltip=folium.GeoJsonTooltip(
        fields=["name", "value"],           # properties에서 가져올 필드
        aliases=["이름", "값"],             # 라벨 텍스트
        localize=True,
        sticky=True,                        # 마우스를 따라다니는지 여부
    ),
    popup=folium.GeoJsonPopup(
        fields=["name", "value"],
        aliases=["이름", "값"],
        localize=True,
        labels=True,
        style="background-color: white;",
    ),
).add_to(m)

folium.LayerControl().add_to(m)

m
````

👉 Choropleth(코로플레스/코로플레스 지도)

“행정구역(구/군/시/도/국가 같은 영역)마다 숫자 값이 있고,
그 숫자 크기를 색 농도나 색깔로 표현하는 지도”

입니다.

기본 단위는 “폴리곤(Polygon, 영역)”
(서울 강남구, 송파구, 종로구… 국경선, 시군구 경계 등)

각 영역마다 하나의 값(인구, 소득, 매출, 센터 수, 측정치 등)이 있음

그 값을 fill_color(색상)으로 표현해서
“어디가 높고, 어디가 낮은지” 한눈에 보여주는 용도

````python
import folium
import pandas as pd
import requests

# 1) 서울 자치구 GeoJSON 가져오기
geo_url = "https://raw.githubusercontent.com/southkorea/seoul-maps/master/kostat/2013/json/seoul_municipalities_geo_simple.json"
res = requests.get(geo_url)
seoul_geo = res.json()

# 2) 예시용 자치구별 지표 데이터 생성
#   - 여기서는 "name"이 GeoJSON의 feature.properties.name_eng과 매칭된다고 가정
data = pd.DataFrame({
    "name": [
        "Gangnam-gu", "Songpa-gu", "Jongno-gu", "Jung-gu",
        "Yongsan-gu", "Mapo-gu", "Gwanak-gu", "Nowon-gu"
    ],
    "value": [95, 88, 70, 72, 80, 78, 65, 60],   # 예시 점수(지표 값)
})

# 3) 지도 생성 (서울시청 근처 중심)
m = folium.Map(location=[37.5665, 126.9780],
               zoom_start=11,
               tiles="cartodbpositron")

# 4) Choropleth 레이어 추가
folium.Choropleth(
    geo_data=seoul_geo,          # GeoJSON 데이터
    name="서울 자치구 지표",
    data=data,                   # Pandas DataFrame
    columns=["name", "value"],   # (키, 값) 컬럼
    key_on="feature.properties.name_eng",  # GeoJSON 매칭 키
    fill_color="YlOrRd",         # 색상 팔레트 (예: YlGnBu, PuRd, BuPu 등)
    fill_opacity=0.7,
    line_opacity=0.3,
    line_color="white",
    nan_fill_color="lightgray",  # 데이터 없는 구는 회색 처리
    legend_name="지표 값(예: 점수, 인구 등)",
    highlight=True,              # 마우스 오버 시 하이라이트
).add_to(m)

# 5) Tooltip/Popup을 GeoJson 레이어로 추가해 더 풍부하게 만들기
folium.GeoJson(
    seoul_geo,
    name="서울 자치구 경계",
    style_function=lambda feature: {
        "fillOpacity": 0,   # Choropleth에서 이미 색을 입혔으니 여기선 투명 (0~1 사이값)
        "color": "black",
        "weight": 0.5,
    },
    tooltip=folium.GeoJsonTooltip(
        fields=["name"],
        aliases=["자치구명"],
        localize=True,
        sticky=True,
    ),
).add_to(m)

folium.LayerControl().add_to(m)

m
````

##### ✔️ MarkerCluster

🟩 개념

**대규모 점 데이터(수백~수만 개)**를 지도에 모두 찍으면 화면이 복잡해지고 렌더링도 느려짐.

MarkerCluster는 지도 축척(zoom level)에 따라 마커들을 자동으로 묶어(clustering) 멀리서 보면 큰 덩어리로

확대하면 개별 마커로자연스럽게 펼쳐지게 해 줌.

👉 주요 옵션

- MarkerCluster(name='레이어명') : 레이어 이름 지정
- disableClusteringAtZoom=n : 특정 zoom 이상에서는 클러스터 해제
- maxClusterRadius=n : 주변 몇 픽셀까지 묶을지(기본 80px)
- spiderfyOnMaxZoom=True/False : 확대 시 마커들이 주변으로 퍼져나오는 효과
- showCoverageOnHover=True/False : 마우스 오버 시 클러스터가 어떤 영역을 커버하는지 보여줌

````python
from folium.plugins import MarkerCluster
import folium

m = folium.Map([37.5665, 126.9780], zoom_start=12)

cluster = MarkerCluster(
    name="매장 위치",
    disableClusteringAtZoom=14,
    maxClusterRadius=60,
    spiderfyOnMaxZoom=True
).add_to(m)

for lat, lon in [
    (37.5665, 126.9780),
    (37.5700, 126.9850),
    (37.5800, 126.9900),
    # ... 실제론 수천개 들어갈 수 있음
]:
    folium.Marker([lat, lon], tooltip="매장").add_to(cluster)

m
````

##### ✔️ HeatMap

🟩 개념

많은 점 데이터를 “점 하나하나 찍는” 대신
**밀도(핫스팟)**를 색으로 시각화.

빨간색=밀집, 파란색/녹색=희박

과학적, 통계적 지도 시각화에서 매우 많이 쓰임.

👉 주요 옵션

- radius : 주변 몇 픽셀까지 영향 줄지 (기본 25)
- blur : 흐림 효과(기본 15)
- gradient : 색상 스펙트럼 지정
- min_opacity : 최소 투명도
- max_zoom : 확대시 표현 방식 조절

````python
from folium.plugins import HeatMap
import folium

m = folium.Map([37.5665, 126.9780], zoom_start=11)

heat_points = [
    [37.5665, 126.9780],
    [37.5700, 126.9820],
    [37.5750, 126.9900],
    # ... 더 많은 좌표
]

HeatMap(
    heat_points,
    radius=20,
    blur=15,
    min_opacity=0.8,
    gradient={0.2: 'blue', 0.5: 'lime', 0.8: 'red'}
).add_to(m)

m
````

````python
import pandas as pd
import numpy as np

@st.cache_data
def get_data():
    df = pd.DataFrame(
        np.random.randn(50, 20), columns=("col %d" % i for i in range(20))
    )
    return df

@st.cache_data
def convert_for_download(df):
    return df.to_csv().encode("utf-8")

df = get_data()
csv = convert_for_download(df)

st.download_button(
    label="Download CSV",
    data=csv,
    file_name="data.csv",
    mime="text/csv",
    icon=":material/download:",
)
````

````python
import folium
from streamlit_folium import st_folium
def create_map():
    locations = [
        [37.5665, 126.9780],  # 서울 시청
        [37.5656, 126.9758],  # 덕수궁
        [37.5682, 126.9777]   # 남대문
    ]
    m = folium.Map(location=locations[0], zoom_start=14)

    for loc in locations:
        folium.Marker(
            location=loc,
            popup=f"위치 {loc}",
            icon=folium.Icon(color='blue', icon='info-sign')
        ).add_to(m)

    return m

st.title("Streamlit + Folium 지도 시각화")
# 지도 생성 및 표시
map_obj = create_map()
st_folium(map_obj, width=800, height=600)
````

### 상태관리 : session_state, caching

#### session_state

> 각 사용자(브라우저 탭)마다 별도 유지되는 상태 dict.

````python
st.title("Session State 예제")

if "count" not in st.session_state:
    st.session_state.count = 0

increment = st.button("증가")
reset = st.button("리셋")
cnt = 0
if increment:
    st.session_state.count += 1
    cnt +=1

if reset:
    st.session_state.count = 0
    cnt = 0
st.write("현재 카운트:", st.session_state.count)
st.write("일반 변수 카운트 :", cnt)
````
