### 조건문 - if 문

의사 결정 작업에 사용되는 제어문 중 하나입니다.

프로그램을 작성하면 위에서부터 아래로 순서대로 실행됩니다. 

하지만 프로그래이 꼭 순서대로만 실행되어서는 올바른 결과를 얻을 수 없습니다. 

경우에 따라서 실행 순서를 변경해야 하는 경우가 생기는데 이때에 조건문을 사용합니다.

<div style='width:100%' >

if 문에 주어진 조건이 true일 때만 실행되는 코드 본문이 포함되어 있습니다.

조건이 false인 경우 다른 조건에 대한 몇 가지 코드가 포함된 선택적 다른 명령문이 실행됩니다.

```python

위에서 부터 명령어가 순차적으로 실행되다가...

if 조건:
    참인 경우 실행할 명령1
    참인 경우 실행할 명령2
    참인 경우 실행할 명령3
    ...

다른 명령1
다른 명령2
다른 명령3
...

```

앞서 if문은 참일때와 거짓일때 나눠서 동작한다고 했습니다.

참이라는것 Bool자료형에서 True를 의미합니다.

if문에서는 `숫자형` , `문자열`, `컬렉션` , `None`과 같은 자료형일때도 동작하는데요.

`숫자형`일 경우에는, 0은 거짓 그외 숫자값이 있으면 참으로 인식합니다.

`문자열`일 경우에는, 빈문자열이면 거짓 그외 1개이상의 문자열이 있으면 참으로 인식합니다.

`컬렉션`은 낯설죠? 우리가 1주차에 배웠었던, 리스트와 튜플, 딕셔너리, set과 같은 자료형을 의미합니다.

이와 같은 자료형은 원소가 없을때는 거짓, 그 외에 자료형 내 원소가 있으면 참으로 인식합니다.

`None`은 처음 보시죠? 아무것도 없다는 자료형을 의미하는데 그래서 if 문에서는 거짓으로 인식합니다.

<span style="font-size:270%">그럼 if문을 같이 사용해볼까요?😁</span> 

````python
a = 10
if a > 20:
    print("정답")
else:
    print("오답")
````

````python
# 숫자형일 경우 숫자값이 있을때, if문 동작을 살펴봅시다.
var = 1

if var:
    print("조건문이 참으로 인식했습니다.")

else: 
    print("조건문이 거짓으로 인식했습니다.")
````

#### 언팩 연산자를 사용하는 튜플 형식의 가변 매개변수 프로그램 작성

````python
def calc_sum(*params):
    total = 0
    for val in params:
        total += val
    return total

ret_val = calc_sum(1,2)
print(f"calc_sum(1, 2) 함수가 반환한 값: {ret_val}")

ret_val = calc_sum(1,2,3)
print(f"calc_sum(1, 2, 3) 함수가 반환한 값: {ret_val}")

ret_val = calc_sum(1,2,3,4)
print(f"calc_sum(1, 2, 3, 4) 함수가 반환한 값: {ret_val}")
````

#### 명시적 매개변수와 가변 매개변수의 혼합 사용 프로그램 작성

````python
def calc_sum(precision, *params):
    if precision == 0:
        total = 0
    elif 0 < precision < 1:
        total = 0.0

    for val in params:
        total += val
    return total

ret_val = calc_sum(0, 1, 2)
print(f"calc_sum(0, 1, 2) 함수가 반환한 값: {ret_val}")

ret_val = calc_sum(0.1, 1, 2)
print(f"calc_sum(0.1, 1, 2) 함수가 반환한 값: {ret_val}")
````

#### 언팩 연사자를 사용하는 튜플 형식의 반환 프로그램 작성

````python
def calc_sum(precision1, precision2, *params):
    if precision1 == 0:
        total1 = 0
    elif 0 < precision1 < 1:
        total1 = 0.0

    if precision2 == 0:
        total2 = 0
    elif 0 < precision2 < 1:
        total2 = 0.0

    for val in params:
        total1 += val
        total2 += val

    return total1, total2


ret_val = calc_sum(0, 0.1, 1, 2)
print("calc_sum(0, 0.1, 1, 2) 함수가 반환한 값: {0}, {1}".format(*ret_val))
print(f"calc_sum(0, 0.1, 1, 2) 함수가 반환한 값: {ret_val[0]}, {ret_val[1]}")
````

#### 매개변수에 함수 전달하는 프로그램

````python
def calc(operator_fn, x, y):
    return operator_fn(x, y)

def plus(op1, op2):
    return op1 + op2

def minus(op1,  op2):
    return op1 - op2

ret_val = calc(plus, 10, 5)
print(f"calc(plus, 10, 5)의 결과 값 : {ret_val}")

ret_val = calc(minus, 10, 5)
print(f"calc(minus, 10, 5)의 결과 값 : {ret_val}")
````

---
#### 람다식 사용하는 프로그램

````python
def calc(operator_fn, x, y):
    return operator_fn(x, y)

ret_val = calc(lambda a, b: a + b, 10, 5)
print(f"calc(lambda a, b: a + b, 10, 5)의 결과 값 : {ret_val}")

ret_val = calc(lambda a, b: a - b, 10, 5)
print(f"calc(lambda a, b: a - b, 10, 5)의 결과 값 : {ret_val}")
````

#### 원의 면적과 둘레 구하는 프로그램 작성

````python
# print("원의 면적: {0:0.2f}, 원의 둘레: {1:0.2f}".format(3.14 * 5 * 5, 2 * 3.14 * 5))

'''
def calc_circle_area(r):
    return 3.14 * r * r

def calc_circumference(r):
    return 2 * 3.14 * r

circle_area = calc_circle_area(5)
circumference = calc_circumference(5)
print(f"원의 면적: {circle_area:0.2f}, 원의 둘레: {circumference:0.2f}")
'''

PI = 3.14

def input_radius():
    radius_str = input("반지름을 입력하세요 : ")
    return float(radius_str)

def calc_circle_area(r):
    return PI * r * r

def calc_circumference(r):
    return 2 * PI * r

radius = input_radius()
circle_area = calc_circle_area(radius)
circumference = calc_circumference(radius)
print(f"원의 면적: {circle_area:0.2f}, 원의 둘레: {circumference:0.2f}")
````
