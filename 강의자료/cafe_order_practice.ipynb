## Phase 1 - 한 번 주문 처리

### 📋 결과 화면 예시

아래와 같이 동작하는 프로그램을 만들어 보세요.

**나의 답:**

### ✏️ 코딩 실습

**[요구사항]**

1. 메뉴는 아메리카노(3000원), 라떼(4000원), 케이크(5000원) 3개로 구성한다.
2. 사용자에게 메뉴 1개를 입력받는다.
3. 수량을 입력받는다.
4. `if / elif / else`를 사용하여 가격을 계산한다.
5. 총 금액을 출력한다.
6. 없는 메뉴를 입력하면 `"없는 메뉴입니다."`를 출력한다.

````python
# Phase 1 - 여기에 코드를 작성하세요
menu = {"아메리카노": 3000, "라떼": 4000, "케이크": 5000}

print("=== 카페 메뉴 ===")
for i in menu:
    print(i)

menu_in = input("메뉴를 입력하세요: ")
if menu_in in menu:
    num_in = input("수량을 입력하세요: ")
    if not num_in.isdigit():
        print("숫자를 입력해야 합니다.")
    else:
        price_in = menu[menu_in]
        final_price = price_in * int(num_in)
        print(f"총금액은 {final_price}원 입니다")
else:
    print('없는 메뉴입니다.')

````

### ✏️ 코딩 실습

**[요구사항]**

1. `for` 문을 사용하여 주문을 **3번 반복**한다.
2. 매번 메뉴와 수량을 입력받는다.
3. 메뉴에 따라 가격을 계산한다.
4. 총 금액을 **누적**한다.
5. 마지막에 총 금액을 출력한다.

````python
# Phase 2 - 여기에 코드를 작성하세요
menu = {"아메리카노": 3000, "라떼": 4000, "케이크": 5000}


print("=== 카페 메뉴 ===")
for i in menu:
    print(i)

final_list =[]
for i in range(3):  # 3번 반복 주문 

    menu_in = input("메뉴를 입력하세요: ")
    if menu_in in menu:
        num_in = input("수량을 입력하세요: ")
        if not num_in.isdigit():
            print("숫자를 입력해야 합니다.")
        else:
            price_in = menu[menu_in]
            final_price = price_in * int(num_in)
            final_list.append(final_price)
            print(f"선택하신 {menu_in} 금액은 {final_price}원 입니다")
    else:
        print('없는 메뉴입니다.')

total_price = sum(final_list)
print(f"총 금액은 {total_price}원 입니다.") 

````
